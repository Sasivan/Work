8. Write, in your own words (a few sentences is enough), the difference between what happens to commit history in a merge versus a rebase, and when you'd choose each — be ready to explain this without notes.


```bash
# A merge preserves both branches' full commit history and joins them with a
# new two-parent commit — nothing is rewritten, history shows the real parallel
# timeline. A rebase replays your branch's commits onto a new base one by one,
# giving each a new hash and producing linear history with no merge commit —
# but it rewrites commit identity, so it's only safe on commits nobody else
# has already pulled.
```
