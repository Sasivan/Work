```python
#with decorator

def Memoize(func):
    cache = {}
    print("wrapper called :")
    def wrapper(n):
        if n not in cache:
            cache[n] = func(n)
        return cache[n]
    wrapper.cache = cache
    return wrapper

@Memoize
def fact(n):
    print("calling...")
    if n <= 1: return 1
    return n * fact(n-1)

print(fact(5))
print(fact(6))
print(fact(4))
print(fact(5))

print(fact.cache) # <------------------------

#In Python, functions are objects, so you can attach your own attributes to them
```

```python
# #Ques 45 -

# Takes a function as input.
# Returns a new function (wrapper).
# Stores previously computed results in a cache.
# Uses the cached result when the same argument is passed again.
# Prints the cache after a few calls.
''' 
@memoize
def fib(...):

is executed once:

memoize(fib)
      │
      ▼
returns wrapper
      │
      ▼
fib = wrapper

After that:

fib(10)
   │
   ▼
wrapper(10)
   │
   ▼
original fib(10)   (only if not cached)
'''

def memoize(func):
    cache = {}
    print("Wrapper called :")
    def wrapper(n):
        if n not in cache:
            cache[n] = func(n)
        return cache[n]
    wrapper.cache = cache
    return wrapper


@memoize
def fib(n):
    print("fib called :",n)
    if n <= 1: return n
    return fib(n-2) + fib(n-1)

print(fib(3))
print(fib(5))
print(fib.cache)
print(fib.cache.values())
```

```python
#Ques 46 -

import random

def retry(func,k):
    last = None
    while k > 0:
        try:
            return func()
        except Exception as e:
            print(f"Attempt : {k} : {e}")
            k -= 1
            last = e
    return last
def guess():
    n = random.randint(1,5)
    if n != 3: raise ValueError(f"Failed : got : {n}")
    else: return "Success"
    
try:
    print(retry(guess,5))
except Exception as e:
    print(f"Last Exception : {e}")
```
