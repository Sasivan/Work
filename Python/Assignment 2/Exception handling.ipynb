```python
#Ques 55 -

def safe_divide(a, b):
    try:
        return a/b
    except ZeroDivisionError:
        print("b can not be zero")
    except TypeError:
        print("enter only numbers")
    finally:
        print("Division attempted")

# Call 1: Valid division
safe_divide(10, 2)

# Call 2: Division by zero
safe_divide(10, 0)

# Call 3: Invalid types
safe_divide("10", 5)
```

```python
#Ques 56 - 

class InsufficientFundsError(Exception):
    def __init__(self, shortfall):
        self.shortfall = shortfall
        super().__init__(f"Insufficient funds. Shortfall amount: {shortfall}")
    

class BankAccount:
    def __init__(self,balance):
        self.balance = balance
    def view(self):
        print(self.balance)
    def deposit(self,amount):
        self.balance += amount
        print("Amount deposited")
    def withdraw(self,amount):
        if self.balance < amount:
            raise InsufficientFundsError(amount - self.balance)
        self.balance -= amount
        print("Amount withdrawn")
        
# Initialize account and perform transactions
account = BankAccount(balance=100)
account.deposit(50)
account.withdraw(40)

# Attempt to withdraw more than the remaining balance (110)
try:
    account.withdraw(150)
except InsufficientFundsError as e:
    print(f"Exception Caught: {e}")
```
