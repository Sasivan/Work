```python
#Ques 32 - 

text = "thisIsACamelCaseString"

#print("".join(' ' + c if c.isupper() else c  for c in text))

lst = []
c = ""
for i in text:
    if i.isupper():
        lst.append(c)
        c = i
    else : c += i
print("_".join(lst).lower())
```

```python
#Ques 33 -

text = list("aaabbbccdddddda")
new_text = ""+text[0]
i = 1;c = 1
while i < len(text):
    if text[i] == text[i-1]:
        c += 1
    else:
        new_text += str(c) + text[i]
        c = 1
    i += 1
print(new_text)
''' 
Always check from back  <-- 
'''
```

```python
#Ques 34 - 
import string
# Test Email 1: Valid
email1 = "user.name@domain.com"

# Test Email 2: Invalid (spaces present)
email2 = "user name@domain.com"

def chech_mail(mail):
    username,domain = mail.split('@')
    if ' ' in username or len(username) < 1:
        print("Invalid Email")
    elif '.' not in domain or domain.startswith(".") or domain.endswith("."):
        print("Invalid Email")
    else: print("Valid Email")
chech_mail(email1)
chech_mail(email2)
```
