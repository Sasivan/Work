```python
#Ques 41 - 

def Grade(score):
    if 90 <= score <= 100: grade = 'A+'
    elif 80 <+ score < 90: grade = 'A'
    elif 70 <= score < 80: grade = 'B'
    elif 60 <= score < 70: grade = 'C'
    elif 50 <= score < 60: grade = 'D'
    else: grade = 'F'
    return grade
def Class(grade):
    if grade in ['A+','A']: print('Distinction')
    elif grade in ['B','C']: print("Pass")
    else: print("Fail")

score = 85.5
grade = Grade(score)
print(grade)
Class(grade)

```

```python
#Ques 42 - 

def valid_triangle(a,b,c):
    return a < b + c and b < a + c and c < a + b
def triangle_type(a,b,c):
    if a == b == c: return "Equilateral"
    elif a == b != c or b == c != a or c == a != b: return "Isosceles"
    else: return "scalene"
def right_angled(a,b,c):
    t = False
    if a**2 == b**2 +c**2: t = True
    elif b**2 == c**2 +a**2: t = True
    elif c**2 == a**2 + b**2: t =True
    return t

a = 3
b = 4
c = 5

print("Valid Triangle : ",valid_triangle(a,b,c))
print("Type : ",triangle_type(a,b,c))
print("Right-Angled : ",right_angled(a,b,c))
```
