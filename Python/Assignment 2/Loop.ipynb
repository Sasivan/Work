```python
#Ques 37 -

import math

lst = []
for i in range(2,500):
    cont = True
    for j in range(2,int(math.sqrt(i)) + 1):
        if i%j ==0 : 
            cont = False
            break
    if cont: lst.append(i)
print(",".join(str(c) for c in lst))
```

```python
#sample code - Selection sort

arr = [29, 10, 14, 37, 13]
n = len(arr)
for i in range(n):
    m = i
    for j in range(i+1,n):
        if arr[m] > arr[j]:
            m = j
    arr[m],arr[i] = arr[i],arr[m]
print(arr)
```
