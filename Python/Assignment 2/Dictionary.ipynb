```python
#Ques 15 -

phone_book = {}
while True:
    print("1 - add contact details")
    print("2 - lookup contact details")
    print("3 - delete contact details")
    print("4 - list all contacts")
    print("5 - Exit")
    choice = input("Enter Choice : ")

    match choice:
        case "1":
            name = input("Enter name : ")
            phone = input("Enter phone : ")
            email = input("Enter email : ")
            phone_book[name] = {"phone":phone,"email":email}
        case "2":
            name = input("Enter name : ")
            for k,v in phone_book.items():
                if k == name: print(phone_book[name])
        case "3":
            name = input("Enter name : ")
            del(phone_book[name])
        case "4": 
            for k,v in phone_book.items():
                print(k,v)
        case "5":
            break
```

```python
#Ques 17 - 

dict1 = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
dict2 = {'b': 2, 'c': 30, 'd': 4, 'e': 5}

data = [[],[],[],[]]
for i,j in zip(dict1.keys(),dict2.keys()):
    if i in dict2 and i not in data[0]: data[0].append(i)
    if j in dict1 and j not in data[0]: data[0].append(j)
    if i in dict1 and i not in dict2: data[1].append(i)
    if j in dict2 and j not in dict1: data[2].append(j)
    if i in dict2 and dict1[i] != dict2[i] and i not in data[3]: data[3].append(i)
    if j in dict1 and dict1[j] != dict2[j] and j not in data[3]: data[3].append(j)
print(data)
#-----------------------------------------------------------------------------------
k1 = dict1.keys()
k2 = dict2.keys()
data[0] = k1 & k2
data[1] = k1 - k2
data[2] = k2 - k1
data[3] = [k for k in data[0] if dict1[k] != dict2[k]]
print(data)
```

```python
#Ques 20 -
import json
json_string = '''
[
    {"name": "Alice", "department": "HR", "salary": 50000},
    {"name": "Bob", "department": "Engineering", "salary": 80000},
    {"name": "Charlie", "department": "HR", "salary": 60000},
    {"name": "David", "department": "Engineering", "salary": 100000}
]
'''

emp = json.loads(json_string)
data = {}
for i in emp:
    dept = i["department"]
    name = i["name"]
    salary = i["salary"]
    if dept not in data:
        data[dept] = {"name":[],"avg_salary":[]}
    data[dept]["name"].append(name)
    data[dept]["avg_salary"].append(salary)
for i,d in data.items():
    d["avg_salary"] = sum(d["avg_salary"])/len(d["avg_salary"])
print(data)
```
