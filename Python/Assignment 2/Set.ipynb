```python
#Ques 26 -

list1 = set([1, 2, 3, 4, 5, 6])
list2 = set([4, 5, 6, 7, 8, 9])

print(f"Union : {list1 | list2}")
print(f"Intersection : {list1 & list2}")
print(f"Difference : {list1 - list2}")
print(f"Symmetric Difference : {list1 ^ list2}")
```

```python
#Ques 27 -

strings = ["apple", "banana", "apple", "cherry", "date", "banana", "fig", "apple"]
new_data = set()
data = set()
for i in strings:
    if i not in data: data.add(i)
    else: new_data.add(i)
print(f"Duplicates : {new_data}")
```
