```python
#Ques 53 - 

filepath = r"C:\Work\Python\Assignment 2\Dataset\data53.csv"

def add():
    d = input("Enter task : ")
    with open(filepath,'a+') as f:
        f.seek(0)
        t = f.read()
        if not t: 
            n = 1 
            f.write(str(n)+' '+d)
        else: 
            n = int(t.split("\n")[-1].split()[0]) + 1
            f.write('\n'+str(n)+' '+d)
def view():
    with open(filepath, "r") as f:
        print(f.read(),end=' ')
        print()
def done():
    d = input("Enter task number: ")
    data = []
    with open(filepath,'r') as f:
        f.seek(0)
        t = f.readlines()
        if d.isdigit():
            for i in t:
                if int(i.split()[0]) != int(d):
                    data.append(i)

    with open(filepath, "w") as f:
        f.writelines(data)

n = 5

while n != 4:
    print()
    print("1 : Add Task")
    print("2 : View Tasks")
    print("3 : Mark done")
    print("4 : Exit")
    print()
    n = int(input("Enter your Choice : "))
    if n == 1:
        add()
    elif n == 2:
        view()
    elif n == 3:
        done()
```
