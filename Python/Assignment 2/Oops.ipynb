```python
#Sample Code - 

# __init__() – Constructor
# __str__() – String representation
# __repr__() – Official representation
# __len__() – len()
# __getitem__() – Indexing
# __setitem__() – Assignment with []
# __contains__() – in
# __iter__() / __next__() – Iterators
# __call__() – Callable objects
# __eq__(), __lt__(), __gt__() – Comparisons
# __add__(), __sub__(), __mul__() – Operator overloading
# __enter__() / __exit__() – Context managers (with statement)

class Box:
    def __init__(self, name, data):
        self.name = name
        self.data = data
        self.i = 0

    # print(obj)
    def __str__(self):
        return f"{self.name} : {self.data}"

    # repr(obj)
    def __repr__(self):
        return f"Box('{self.name}',{self.data})"

    # len(obj)
    def __len__(self):
        return len(self.data)

    # obj[index]
    def __getitem__(self, index):
        return self.data[index]

    # obj[index] = value
    def __setitem__(self, index, value):
        self.data[index] = value

    # value in obj
    def __contains__(self, value):
        return value in self.data

    # iterator
    def __iter__(self):
        self.i = 0
        return self

    # next(iterator)
    def __next__(self):
        if self.i >= len(self.data):
            raise StopIteration
        x = self.data[self.i]
        self.i += 1
        return x

    # obj()
    def __call__(self):
        print("Object Called")

    # ==
    def __eq__(self, other):
        return len(self.data) == len(other.data)

    # <
    def __lt__(self, other):
        return len(self.data) < len(other.data)

    # >
    def __gt__(self, other):
        return len(self.data) > len(other.data)

    # +
    def __add__(self, other):
        return self.data + other.data

    # -
    def __sub__(self, other):
        return len(self.data) - len(other.data)

    # *
    def __mul__(self, n):
        return self.data * n

    # with statement
    def __enter__(self):
        print("Enter")
        return self

    def __exit__(self, exc_type, exc, tb):
        print("Exit")


a = Box("A", [1, 2, 3])
b = Box("B", [4, 5])

print(a)                  # __str__
print(repr(a))            # __repr__

print(len(a))             # __len__

print(a[1])               # __getitem__

a[1] = 10                 # __setitem__
print(a)

print(10 in a)            # __contains__

for i in a:               # __iter__ __next__
    print(i)

a()                       # __call__

print(a == b)             # __eq__
print(a < b)              # __lt__
print(a > b)              # __gt__

print(a + b)              # __add__
print(a - b)              # __sub__
print(a * 2)              # __mul__

with a as obj:            # __enter__ __exit__
    print(obj)
```

```python
#Ques 57 - 

class Rectangle():
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def area(self):
        return self.a * self.b
    def perimeter(self):
        return 2*(self.a+self.b)
    def is_square(self):
        return self.a == self.b
    def __str__(self):
        return f"Rectangle - {self.a,self.b}"
# Create instances
rect1 = Rectangle(4,5)
rect2 = Rectangle(3,3)

# Print descriptions and calculate properties
print(rect1)
print(f"Area: {rect1.area()} | Perimeter: {rect1.perimeter()} | Square: {rect1.is_square()}")

print(rect2)
print(f"Area: {rect2.area()} | Perimeter: {rect2.perimeter()} | Square: {rect2.is_square()}")
```

```python
#Ques 58 - 

class Employee:
    def __init__(self,name,role,salary):
        self.name = name
        self.role = role
        self.salary = salary
    def display(self):
        print(f"Name: {self.name}")
        print(f"Role: {self.role}")
        print(f"Salary: {self.salary}")
class Manager(Employee):
    def __init__(self, name, role, salary):
        super().__init__(name, role, salary)
        self.team = []
    def add(self, employee):
        self.team.append(employee)
    def calc(self):
        total = self.salary

        for employee in self.team:
            total += employee.salary

        return total
    def __str__(self):
        return f"{self.name} ({self.role})"

# Create basic employees
emp1 = Employee(name="Alice", role="Developer", salary=80000)
emp2 = Employee(name="Bob", role="QA aa_to_teamEngineer", salary=70000)

# Create manager and build team
mgr = Manager(name="Charlie", role="Engineering Manager", salary=130000)
mgr.add(emp1)
mgr.add(emp2)

# Display information
print(mgr)
print(f"Total Team Salary Cost: ${mgr.calc()}")
```

```python
#Ques 60 -
from datetime import datetime

class logs:
    def __init__(self,info,date,msg):
        self.info = info
        self.date = date
        self.msg = msg
    def __str__(self):
        return f"{self.info} {self.date} {self.msg} \n"   
    __repr__ = __str__
class LogAnalyzer:
    def __init__(self,log):
        self.log = log
        self.d = []
    def prep(self):
        for i in self.log:
            data = i.split(' ')
            self.d.append(logs(data[0],datetime.strptime(data[1], "%Y-%m-%d").date(),' '.join(data[2::])))
    def filter_by_level(self,level):
        for i in self.d:
            if i.info == f"[{level}]":
                print(i)
    def count_by_level(self):
        data = {"[ERROR]":0,"[INFO]":0,"[WARNING]":0}
        for i in self.d:
            if i.info not in data.keys():
                data[i.info] = 1
            else:
                data[i.info] += 1
        print(data)
    def find_by_date(self,Date):
        lst = []
        for i in self.d:
            if i.date == datetime.strptime(Date, "%Y-%m-%d").date():
                lst.append(i)
        return lst
    def get_recent(self,n):
        return self.d[-n:]

    def __str__(self):
        return f"{self.d}"

log = [
    "[ERROR] 2024-01-15 Database connection failed",
    "[INFO] 2024-01-15 Server started on port 8080",
    "[WARNING] 2024-01-16 Memory usage at 85%",
    "[ERROR] 2024-01-17 Disk space critical"
]

analyzer = LogAnalyzer(log)
analyzer.prep()
print(analyzer)
print()
analyzer.filter_by_level("ERROR")
analyzer.count_by_level()
print("Jan 15 Logs:", analyzer.find_by_date("2024-01-15"))
print("Recent 2 Logs:", analyzer.get_recent(2))
```

```python
#Ques 61 - 
from abc import ABC,abstractmethod
import math
class Shape(ABC):
    def __init__(self):
        pass
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def describe(self):
        pass
class Circle(Shape):
    def __init__(self,r):
        self.r = r
    def area(self):
        return math.pow(self.r,2) * math.pi
    def describe(self):
        return f"Shape : Circle with radius {self.r}"
class Triangle(Shape):
    def __init__(self,b,h):
        self.b = b
        self.h = h
    def area(self):
        return self.b * self.h * .5
    def describe(self):
        return f"Shape : Triangle with base {self.b} | height {self.h}"
class Trapezoid(Shape):
    def __init__(self,b1,b2,h):
        self.b1 = b1
        self.b2 = b2
        self.h = h
    def area(self):
        return (self.b1+self.b2) * self.h * .5
    def describe(self):
        return f"Shape : Trapezoid with base1 {self.b1} | base2 {self.b2} | height {self.h}"

def show_shapes(shapes):
    area = 0
    for shape in shapes:
        print(shape.describe())
        area += shape.area()
        print("Area:", shape.area())
    print(f"Total area : {area:.2f} ")
# Instantiate mixed shapes
shapes = [
    Circle(5),
    Triangle(4, 6),
    Trapezoid(3, 5,4)
]

show_shapes(shapes)
```
