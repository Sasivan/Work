```python
#Ques 22 -
numbers = (12, 7, 3, 20, 15, 8, 4, 11)
even = ()
odd = ()
for i in numbers:
    if i%2==0: even+=(i,)
    else: odd += (i,)
print(f"even : {even}\nodd : {odd}")
```

```python
#Ques 23 -

coordinates = [(1.0, 2.0), (4.0, 6.0), (7.0, 2.0)]
d = []
for i in range(len(coordinates) - 1):
    x1 = coordinates[i][0]
    y1 = coordinates[i][1]
    x2 = coordinates[i+1][0]
    y2 = coordinates[i+1][1]
    
    dis = ((x2-x1)**2+(y2-y1)**2)**0.5
    print((x1,y1)," --- ",(x2,y2)," : ",dis)
    d.append(dis)
print(d)
```

```python
#Ques 24 - 

mixed_tuple = (10, "hello", 3.14, True, 42, "world", False, 2.71)

dct = {"int":(),"str":(),"float":(),"bool":()}

for i in mixed_tuple:
    if isinstance(i,int): dct["int"] += (i,)
    if isinstance(i,str): dct["str"] += (i,)
    if isinstance(i,float): dct["float"] += (i,)
    if isinstance(i,bool): dct["bool"] += (i,)
print(dct)
    
```

```python
#Ques 25 -

def calc(data):
    return (sum(data),sum(data)/len(data),max(data),min(data))

data = [10, 20, 30, 40, 50]

Sum,Avg,Max,Min = calc(data)
print(f"Sum : {Sum} Avg : {Avg:.1f} Max : {Max} Min : {Min}")

```
