```python
# ==========================================
# 1. LISTS (Selected: Q6)
# ==========================================
# Question 6:
# Write a program that rotates a list by n positions to the right. 
# For example, rotating [1, 2, 3, 4, 5] by 2 gives [4, 5, 1, 2, 3]. 
# Handle cases where n is greater than the list length.
#
# Sample Input:  lst = [1, 2, 3, 4, 5], n = 2
# Sample Output: [4, 5, 1, 2, 3]

def rotate_list(lst, n):
    return lst[-n:]+lst[:-n]
    # return lst[len(lst)-n:]+lst[:len(lst) - n
if __name__ == '__main__':
    # Test Question 6
    sample_list = [1, 2, 3, 4, 5]
    print("Q6 Output:", rotate_list(sample_list, 2))
```

```python
# ==========================================
# 2. DICTIONARIES (Selected: Q12)
# ==========================================
# Question 12:
# Given two dictionaries, write a program that merges them. 
# If a key exists in both, add their values together (assume all values are numeric).
#
# Sample Input:  dict1 = {'a': 100, 'b': 200, 'c': 300}
#                dict2 = {'b': 150, 'c': 50, 'd': 400}
# Sample Output: {'a': 100, 'b': 350, 'c': 350, 'd': 400}

def merge_dictionaries(dict1, dict2):
    for i in dict1.keys():
        if i in dict2.keys():
            dict1[i] += dict2[i]
    for i in dict2.keys():
        if i not in dict1.keys():
            dict1[i] = dict2[i]
    return dict1 

if __name__ == '__main__':
    # Test Question 12
    d1 = {'a': 100, 'b': 200, 'c': 300}
    d2 = {'b': 150, 'c': 50, 'd': 400}
    print("Q12 Output:", merge_dictionaries(d1, d2))
```

```python
# ==========================================
# 4. SETS (Selected: Q28)
# ==========================================
# Question 28:
# A school offers three elective courses. You are given three sets, each containing 
# student names enrolled in that course. Write a program that finds: 
# (a) students enrolled in all three courses, 
# (b) students enrolled in exactly one course, and 
# (c) students enrolled in at least two courses.
#
# Sample Input:  math_class = {"Alice", "Bob", "Charlie", "Diana"}
#                science_class = {"Bob", "Charlie", "Evan", "Fiona"}
#                history_class = {"Charlie", "Diana", "Fiona", "George"}
# Sample Output: All Three: {'Charlie'}
#                Exactly One: {'Alice', 'Evan', 'George'}
#                At Least Two: {'Bob', 'Charlie', 'Diana', 'Fiona'}

def analyze_course_enrollments(math_class, science_class, history_class):
    all3 = math_class & science_class & history_class
    only1 = ((math_class-science_class-history_class)|
             (science_class-history_class-math_class)|
             (history_class-math_class-science_class))
    atleast2 = (math_class&science_class)| (science_class&history_class) | (history_class&math_class)
    return (all3,only1,atleast2)
if __name__ == '__main__':
    # Test Question 28
    m = {"Alice", "Bob", "Charlie", "Diana"}
    s = {"Bob", "Charlie", "Evan", "Fiona"}
    h = {"Charlie", "Diana", "Fiona", "George"}
    print("Q28 Output:", analyze_course_enrollments(m, s, h))
```

```python
# ==========================================
# 5. STRINGS (Selected: Q32)
# ==========================================
# Question 32:
# Write a program that takes a CamelCase string (e.g., “thisIsACamelCaseString”) 
# and converts it to snake_case (e.g., “this_is_a_camel_case_string”).
#
# Sample Input:  "thisIsACamelCaseString"
# Sample Output: "this_is_a_camel_case_string"

def camel_to_snake(text):
    res = ""
    for i in text:
        if i.isupper():
            res += "-"+i.lower()
        else:
            res += i
    return res
if __name__ == '__main__':
    # Test Question 32
    print("Q32 Output:", camel_to_snake("thisIsACamelCaseString"))
```

```python
# ==========================================
# 6. LOOPS — FOR AND WHILE (Selected: Q38)
# ==========================================
# Question 38:
# Write a program using a for loop that generates the first n terms of the Fibonacci 
# sequence and stores them in a list. Then use another loop to print each term alongside its index.
#
# Sample Input:  n = 6
# Sample Output: (1, 0) (2, 1) (3, 1) (4, 2) (5, 3) (6, 5)

def generate_fibonacci_terms(n):
    data = [0,1,1]
    if n <=2: return enumerate(data[:n])
    for i in range(3,n+1):
        data.append(data[i-1]+data[i-2])
    return [(i+1,data[i]) for i in range(n)]

if __name__ == '__main__':
    # Test Question 38
    print("Q38 Output:", generate_fibonacci_terms(6))
```

```python
# ==========================================
# 7. CONDITIONAL STATEMENTS (Selected: Q41)
# ==========================================
# Question 41:
# Write a program that takes a student’s percentage score and prints the corresponding grade 
# using this scale: 90–100 -> A+, 80–89 -> A, 70–79 -> B, 60–69 -> C, 50–59 -> D, below 50 -> F. 
# Also print “Distinction” if the grade is A+ or A, “Pass” if C or above, and “Fail” otherwise.
#
# Sample Input:  score = 85
# Sample Output: Grade: A, Status: Distinction

def calculate_student_grade(score):
    grade = ''
    if 100 >= score >= 90:  grade = "A+"
    elif 90 > score >= 80:  grade = "A"
    elif 80 > score >= 70:  grade = "B"
    elif 70 > score >= 60:  grade = "C"
    elif 60 > score >= 50:  grade = "D"
    else: grade = "F"

    if grade in ["A+","A"]: return (grade, "Distinction")
    elif grade in ["B","C"]: return (grade,"Pass")
    else: return (grade,"fail")

if __name__ == '__main__':
    # Test Question 41
    print("Q41 Output:", calculate_student_grade(85))
```

```python
# ==========================================
# 8. FUNCTIONS (Selected: Q45)
# ==========================================
# Question 45:
# Write a function memoize that takes another function as input and returns a new version 
# of that function which caches results. If the function is called again with the same 
# arguments, it should return the cached result instead of recomputing. Test it with a 
# recursive Fibonacci function and print the cache contents after several calls.
#
# Sample Output: fib(5) = 5
#                Cache: {1: 1, 0: 0, 2: 1, 3: 2, 4: 3, 5: 5}

def memoize(func):
    cache = {}
    def wrapper(n):
        if n not in cache.keys():
            cache[n] = func(n)
        return cache[n]
    wrapper.cache = cache
    return wrapper

if __name__ == '__main__':
    # Test Question 45
    @memoize
    def fib(n):
        if n <= 1: return n
        return fib(n - 1) + fib(n - 2)
    
    if fib(5) is not None:
        print("Q45 Output Fib Result:", fib(5))
        print("Q45 Cache Contents:", getattr(fib, 'cache', {}))
```

```python
# ==========================================
# 9. LIST COMPREHENSIONS (Selected: Q49)
# ==========================================
# Question 49:
# Given a list of file names like ['report.pdf', 'data.csv', 'image.png', 'notes.txt', 
# 'summary.pdf', 'photo.jpg'], use a list comprehension to extract only the file names 
# (without extensions) of .pdf and .csv files. Result should be ['report', 'data', 'summary'].
#
# Sample Input:  files = ['report.pdf', 'data.csv', 'image.png', 'notes.txt', 'summary.pdf', 'photo.jpg']
# Sample Output: ['report', 'data', 'summary']

def extract_pdf_csv_names(files):
    return [x.split('.')[0] for x in files]

if __name__ == '__main__':
    # Test Question 49
    sample_files = ['report.pdf', 'data.csv', 'image.png', 'notes.txt', 'summary.pdf', 'photo.jpg']
    print("Q49 Output:", extract_pdf_csv_names(sample_files))
```
