````python
# ==========================================
# 1. LISTS
# ==========================================
# Question 1:
# Write a function called move_zeros_to_end that takes a list of integers and 
# moves all zero values to the end of the list while maintaining the relative 
# order of all non-zero elements. Perform this operation in-place without using 
# extra list memory allocations.
#
# Sample Input:  nums = [0, 1, 0, 3, 12, 0, 5]
# Sample Output: [1, 3, 12, 5, 0, 0, 0]

def move_zeros_to_end(nums):
    for i in nums:
        if i == 0:
            nums.remove(i)
            nums.append(0)
    return nums

if __name__ == '__main__':
    # Test Question 1
    sample_nums = [0, 1, 0, 3, 12, 0, 5]
    move_zeros_to_end(sample_nums)
    print("Q1 Output:", sample_nums)
````

````python
# ==========================================
# 2. DICTIONARIES
# ==========================================
# Question 2:
# Write a function called aggregate_inventory that takes a list of order 
# dictionaries (each with keys 'item', 'quantity', 'unit_price') and returns a 
# summary dictionary. The summary dictionary should map each 'item' to a nested 
# dictionary containing its total 'total_quantity' and total 'total_revenue'.
#
# Sample Input:  orders = [
#                    {'item': 'Laptop', 'quantity': 2, 'unit_price': 1000},
#                    {'item': 'Mouse', 'quantity': 5, 'unit_price': 20},
#                    {'item': 'Laptop', 'quantity': 1, 'unit_price': 1000}
#                ]
# Sample Output: {
#                    'Laptop': {'total_quantity': 3, 'total_revenue': 3000},
#                    'Mouse': {'total_quantity': 5, 'total_revenue': 100}
#                }

def aggregate_inventory(orders):
    res = {}
    for k in orders:
        if k["item"] not in res.keys():
            res[k["item"]] = {'total_quantity': k["quantity"], 'total_revenue': k["unit_price"]}
        else:
            res[k["item"]]["total_quantity"] += k["quantity"]
            res[k["item"]]["total_revenue"] += k["unit_price"]
    return res
if __name__ == '__main__':
    # Test Question 2
    sample_orders = [
        {'item': 'Laptop', 'quantity': 2, 'unit_price': 1000},  
        {'item': 'Mouse', 'quantity': 5, 'unit_price': 20},
        {'item': 'Laptop', 'quantity': 1, 'unit_price': 1000}
    ]
    print("Q2 Output:\n", aggregate_inventory(sample_orders))
````

````python
# ==========================================
# 3. TUPLES
# ==========================================
# Question 3:
# Given a list of employee records stored as tuples in the format 
# (id, name, department, (year, month, day)), write a function called filter_early_hires 
# that extracts and returns a list of employee names who were hired before January 1, 2022. 
# Use tuple unpacking to access the nested join date tuple cleanly.
#
# Sample Input:  employees = [
#                    (101, "Alice", "HR", (2020, 5, 12)),
#                    (102, "Bob", "Engineering", (2023, 1, 15)),
#                    (103, "Charlie", "Finance", (2021, 11, 30))
#                ]
# Sample Output: ["Alice", "Charlie"]

def filter_early_hires(employees):
    ot = []
    for i in employees:
        if i[3][0] <= 2022 or i[3][1] <= 1 and i[3][2] <=1:
            ot.append(i[1])
    return ot

if __name__ == '__main__':
    # Test Question 3
    sample_employees = [
        (101, "Alice", "HR", (2020, 5, 12)),
        (102, "Bob", "Engineering", (2023, 1, 15)),
        (103, "Charlie", "Finance", (2021, 11, 30))
    ]
    print("Q3 Output:", filter_early_hires(sample_employees))
````

````python
# ==========================================
# 5. STRINGS
# ==========================================
# Question 5:
# Write a function called extract_query_params(url) that takes a standard URL 
# string containing query parameters and extracts the parameters into a key-value dictionary. 
# Do not use any external or built-in parsing libraries (like urllib) — use string methods.
#
# Sample Input:  url = "https://shop.com/items?category=books&status=active&page=1"
# Sample Output: {'category': 'books', 'status': 'active', 'page': '1'}

def extract_query_params(url):
    ot = {}
    url = url.split("?")[1].split("&")
    for i in url:
        o = i.split("=")
        ot[o[0]] = o[1]
    return ot


if __name__ == '__main__':
    # Test Question 5
    sample_url = "https://shop.com/items?category=books&status=active&page=1"
    print("Q5 Output:", extract_query_params(sample_url))
````

````python
# ==========================================
# 6. LOOPS (FOR & WHILE)
# ==========================================
# Question 6:
# Write a function called max_profit that processes an array representing stock prices over time 
# using a loop. Track and return the maximum profit achievable from buying on one day and selling 
# on a future day. If no profit is possible, return 0.
#
# Sample Input:  prices = [7, 1, 5, 3, 6, 4]
# Explanation:   Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6 - 1 = 5.
# Sample Output: 5

def max_profit(prices):
    Max = 0
    i = 0
    while i < len(prices):
        j = i+1
        while j < len(prices):
            Max = max(prices[j] - prices[i],Max)
            j += 1
        i +=1
    return Max


if __name__ == '__main__':
    # Test Question 6
    sample_prices = [7, 1, 5, 3, 6, 4]
    print("Q6 Output:", max_profit(sample_prices))
````

````python
# ==========================================
# 8. FUNCTIONS
# ==========================================
# Question 8:
# Write a higher-order function called create_rate_limiter(max_calls) that returns a 
# closure function. The returned function should track how many times it has been called. 
# If called within max_calls, it executes a passed callback function; otherwise, it returns 
# "Rate limit exceeded!".
#
# Sample Usage:
# limiter = create_rate_limiter(max_calls=2)
# limiter(lambda: "Data fetched")  --> Output: "Data fetched"
# limiter(lambda: "Data fetched")  --> Output: "Data fetched"
# limiter(lambda: "Data fetched")  --> Output: "Rate limit exceeded!"

def create_rate_limiter(max_calls):
    call =0
    def wrapper(s):
        nonlocal call
        if call < max_calls:
            call += 1
            return s()
        else:
            return "Rate limit exceeded!"
    return wrapper

if __name__ == '__main__':
    # Test Question 8
    limiter = create_rate_limiter(max_calls=2)
    if limiter:
        print("Q8 Call 1:", limiter(lambda: "Data fetched"))
        print("Q8 Call 2:", limiter(lambda: "Data fetched"))
        print("Q8 Call 3:", limiter(lambda: "Data fetched"))
````

````python
# ==========================================
# 9. LIST COMPREHENSIONS
# ==========================================
# Question 9:
# Write a function called filter_top_students that takes a dictionary mapping student names 
# to lists of assignment scores. Return a sorted list of names for all students whose average 
# score is greater than or equal to 75.0 using a single list comprehension.
#
# Sample Input:  grades = {
#                    'Alice': [80, 90, 85],
#                    'Bob': [60, 70, 65],
#                    'Charlie': [70, 80, 75]
#                }
# Sample Output: ['Alice', 'Charlie']

def filter_top_students(grades):
    return [k for k,v in grades.items() if sum(v)/len(v) >= 75]

if __name__ == '__main__':
    # Test Question 9
    sample_grades = {
        'Alice': [80, 90, 85],
        'Bob': [60, 70, 65],
        'Charlie': [70, 80, 75]
    }
    print("Q9 Output:", filter_top_students(sample_grades))
````

````python
# ==========================================
# 11. EXCEPTION HANDLING
# ==========================================
# Question 11:
# Define a custom exception DataValidationError. Then write a function safe_json_parser 
# that accepts a JSON string and required_keys list. Handle JSONDecodeError, KeyError, and TypeError, 
# raising DataValidationError with descriptive error messages.
import json

class DataValidationError(Exception):
    pass

def safe_json_parser(json_string, required_keys):
    try:
        data = json.loads(json_string)
        if not isinstance(data,dict):
            raise TypeError("JSON must contain an object")
        for i in required_keys:
            if i not in data.keys():
                raise KeyError(i)
        return data
    except TypeError as e:
        raise DataValidationError(e)
    except KeyError as e:
        raise DataValidationError(f"Missing key : {e}")
    except json.JSONDecodeError as e:
        raise DataValidationError("Invalid JSON format")


if __name__ == '__main__':
    # Test Question 11
    try:
        # 1. Valid JSON
        safe_json_parser('{"name":"Alice","age":25}', ["name", "age"])

        # 2. Missing key
        safe_json_parser('{"name":"Alice"}', ["name", "age"])

        # 3. Invalid JSON
        safe_json_parser('{name:"Alice"}', ["name"])

        # 4. JSON array (TypeError)
        safe_json_parser('["Alice","Bob"]', ["name"])

        # 5. Empty JSON object
        safe_json_parser('{}', ["id"])
    except DataValidationError as e:
        print("Q11 Output Caught Error:", e)
    except Exception:
        print("Q11: Function ready for implementation.")
````

````python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (OOP)
# ==========================================
# Question 12:
# Create a base class Vehicle (make, model, rental_rate_per_day). Create subclasses Car 
# (seating_capacity) and Truck (cargo_capacity_tons). Override calculate_rental_cost(days) 
# where Truck adds a $50 flat surcharge per ton of cargo. Write a polymorphic function process_rentals.

class Vehicle:
    def __init__(self, make, model, rental_rate_per_day):
        self.make = make 
        self.model = model
        self.rental_rate_per_day = rental_rate_per_day

class Car(Vehicle):
    def __init__(self, make, model, rental_rate_per_day,seating_capacity):
        self.seating_capacity = seating_capacity
        super().__init__(make, model, rental_rate_per_day)
    def cost(self,d):
            return self.rental_rate_per_day*d
class Truck(Vehicle):
    def __init__(self, make, model, rental_rate_per_day,cargo_capacity_tons):
        self.cargo_capacity_tons = cargo_capacity_tons
        super().__init__(make, model, rental_rate_per_day)
    def cost(self,d):
        return (self.rental_rate_per_day*d) + (50*self.cargo_capacity_tons)
        

def process_rentals(vehicles, days):
    for i in vehicles:
        print(i.cost(days))

if __name__ == '__main__':
    # Test Question 12
    car1 = Car("Toyota", "Corolla", 50, 5)
    car2 = Car("Honda", "City", 60, 4)

    truck1 = Truck("Volvo", "FH16", 120, 3)
    truck2 = Truck("Tata", "Signa", 100, 2)

    vehicles = [car1, car2, truck1, truck2]

    process_rentals(vehicles, 5)
````
