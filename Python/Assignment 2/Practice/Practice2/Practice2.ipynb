```python
# ==========================================
# 1. LISTS
# ==========================================
# Question 1:
# Write a function called filter_above_average that takes a list of numbers, 
# calculates its average, and returns a new list containing only the numbers 
# that are strictly greater than or equal to the average.
#
# Sample Input:  numbers = [10, 25, 40, 5, 15]
# Sample Output: [25, 40]

def filter_above_average(lst):
    avg = sum(lst)/len(lst)
    return list(filter(lambda x: x >= avg,lst))
numbers = [10, 25, 40, 5, 15]
print(filter_above_average(numbers))
```

```python
# ==========================================
# 2. DICTIONARIES
# ==========================================
# Question 2:
# Write a program that takes a list of lowercase words and creates a dictionary 
# where each key is a starting letter and the corresponding value is a list of 
# all words from the input starting with that letter.
#
# Sample Input:  words = ["apple", "apricot", "banana", "berry", "cherry"]
# Sample Output: {'a': ['apple', 'apricot'], 'b': ['banana', 'berry'], 'c': ['cherry']}

words = ["apple", "apricot", "banana", "berry", "cherry"]
dict1 = {}
for i in words:
    if i[0] not in dict1.keys():
        dict1[i[0]] = [i]
    else:
        dict1[i[0]].append(i)
print(dict1)
```

```python
# ==========================================
# 3. TUPLES
# ==========================================
# Question 3:
# Given a list of 2D coordinate tuples (x, y), write a program to find 
# the point closest to the origin (0, 0) using Python's min() function 
# and math.hypot.
#
# Sample Input:  points = [(3, 4), (1, 2), (5, 12), (-1, -1)]
# Sample Output: Closest Point: (-1, -1)

import math
points = [(3, 4), (1, 2), (5, 12), (-1, -1)]
dis = {}
for i in points:
    dis[i] = math.hypot(i[0],i[1])
print(list(filter(lambda x: dis[x] == min(dis.values()),dis.keys())))
```

```python
# ==========================================
# 4. SETS
# ==========================================
# Question 4:
# Given three string variables, write a program using set intersection 
# to find all unique lowercase characters that appear in all three strings.
#
# Sample Input:  s1 = "notebook", s2 = "textbook", s3 = "scrapbook"
# Sample Output: {'o', 'k', 'b'}

s1 = "notebook"; s2 = "textbook"; s3 = "scrapbook"
S1 = set([x for x in s1])
S2 = set([x for x in s2])
S3 = set([x for x in s3])
print(set(S1&S2&S3))
print(set(s1)&set(s2)&set(s3))
```

```python
# ==========================================
# 5. STRINGS
# ==========================================
# Question 5:
# Write a function mask_sensitive_data(card_num) that takes a credit card 
# or phone number string and masks all characters except the last 4 digits 
# with asterisks ('*').
#
# Sample Input:  card_num = "9876543210984321"
# Sample Output: "************4321"

def mask_sensitive_data(txt):
    return f"{"*"*(len(txt)-4)}{txt[-4:]}"
print(mask_sensitive_data("9876543210984321"))
```

```python
# ==========================================
# 6. LOOPS (FOR & WHILE)
# ==========================================
# Question 6:
# Write a program using nested loops to print an inverted right-angled 
# number pyramid for a given positive integer n.
#
# Sample Input:  n = 5
# Sample Output:
# 5 5 5 5 5
# 4 4 4 4
# 3 3 3
# 2 2
# 1

for i in range(5,0,-1):
    for j in range(5,0,-1):
        if i >= j: 
            print(i,end=" ")
    print()
```

```python
# ==========================================
# 7. CONDITIONAL STATEMENTS
# ==========================================
# Question 7:
# Write a program that accepts three side lengths (a, b, c). First verify 
# if they form a valid triangle (a + b > c, a + c > b, b + c > a). If valid, 
# classify it as "Equilateral", "Isosceles", or "Scalene".
#
# Sample Input 1: a = 3, b = 3, c = 3   --> "Equilateral"
# Sample Input 2: a = 5, b = 5, c = 8   --> "Isosceles"
# Sample Input 3: a = 1, b = 2, c = 10  --> "Invalid Triangle"

def check(a,b,c):
    if a+b<c or b+c<a or a+c<b:
        print("Invalid")
    elif a == b and b == c and c==a:
        print("Equilateral")
    elif a==b and b!= c or a==c and c!=a or b==c and b!=a:
        print("Isosceles")
    else:
        print("Valid")

check(3,3,3)
check(5,5,8)
check(1,2,10)
```

```python
# ==========================================
# 8. FUNCTIONS
# ==========================================
# Question 8:
# Write a recursive function digital_root(n) that repeatedly calculates 
# the sum of the digits of a non-negative integer until a single-digit 
# number is obtained.
#
# Sample Input:  9875
# Explanation:   9 + 8 + 7 + 5 = 29  -->  2 + 9 = 11  -->  1 + 1 = 2
# Sample Output: 2



def digital_root(n):
    while n >= 10:
        s = 0
        while n > 0:
            s += n%10
            n = n//10
        n = s
    return n
print(digital_root(9875))
```

```python
# ==========================================
# 9. LIST COMPREHENSIONS
# ==========================================
# Question 9:
# Given a 2D list of positive and negative integers (matrix), use a single 
# list comprehension to flatten the matrix into a 1D list containing only 
# the positive even numbers.
#
# Sample Input:  matrix = [[-2, 3, 4], [8, -5, 6], [0, 10, -12]]
# Sample Output: [4, 8, 6, 10]

matrix = [[-2, 3, 4], [8, -5, 6], [0, 10, -12]]
print( [x for y in matrix for x in y  if x > 0 and x % 2 == 0])
```

```python
# ==========================================
# 10. FILE HANDLING
# ==========================================
# Question 10:
# Write a function replace_word_in_file(filepath, target, replacement) 
# that opens a text file, replaces all occurrences of the target word 
# with replacement, and saves the updated text back to the file.
#
# Sample Input File Content: "Python is great. Python is easy."
# Function Call: replace_word_in_file("test.txt", "Python", "Java")
# Updated File Content:      "Java is great. Java is easy."

def replace_word_in_file(filepath,a,b):
    with open(filepath,'r+') as f:
        data = f.read()
        data = data.replace(a,b)
        f.seek(0)
        f.write(data)

replace_word_in_file(r"C:\Work\Python\Assignment 2\Dataset\test.txt", "Python", "Java")
```

```python
# ==========================================
# 11. EXCEPTION HANDLING
# ==========================================
# Question 11:
# Write a function get_element(lst, index, default=None) that attempts 
# to access lst[index]. Handle both IndexError and TypeError gracefully 
# using try-except, returning the default value if an error occurs.
#
# Sample Input 1: get_element([10, 20, 30], 1)     --> Output: 20
# Sample Input 2: get_element([10, 20, 30], 5, -1)  --> Output: -1
# Sample Input 3: get_element([10, 20, 30], "a", 0) --> Output: 0

def get_element(lst, index, default=None):
    try:
        return lst[index]
    except (TypeError,IndexError):
        return default

print("Output 1:", get_element([10, 20, 30], 1))
print("Output 2:", get_element([10, 20, 30], 5, -1))
print("Output 3:", get_element([10, 20, 30], "a", 0))
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (OOP)
# ==========================================
# Question 12:
# Create a BankAccount class initialized with account_number, holder_name, 
# and balance. Implement a method transfer(self, target_account, amount) 
# that deducts amount from the current account and deposits it into 
# target_account only if sufficient balance is available.

class BankAccount:
    def __init__(self,account_number,holder_name,balance):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance
    def transfer(self, target_account, amount):
        if self.balance < amount:
            return "false"
        self.balance -= amount
        target_account.balance += amount

acc1 = BankAccount("A101", "Alice", 500)
acc2 = BankAccount("B202", "Bob", 200)
acc1.transfer(acc2, 150)

print("Output:")
print(f"{acc1.holder_name} Balance: {acc1.balance}")
print(f"{acc2.holder_name} Balance: {acc2.balance}")
```
