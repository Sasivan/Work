## Q1 — OOP with Abstract Base Classes

```python
from abc import ABC, abstractmethod

class Appliance(ABC):
    @abstractmethod
    def power_consumption(self):
        pass
    def label(self):
        return f"This appliance uses {self.power_consumption()} watts"

class AC(Appliance):
    def __init__(self, tonnage):
        self.tonnage = tonnage
    def power_consumption(self):
        return self.tonnage * 1500

class Fridge(Appliance):
    def __init__(self, capacity_litres):
        self.capacity_litres = capacity_litres
    def power_consumption(self):
        return self.capacity_litres * 2

appliances = [AC(1.5), Fridge(300), AC(2)]
for a in appliances:
    print(a.label())

try:
    Appliance()
except TypeError as e:
    print("Confirmed abstract class cannot be instantiated:", e)
```

## Q2 — Generators & File I/O

```python
with open("sensor_log.txt", "w") as f:
    f.write("\n".join(["23.5","24.1","99.9","25.0","-5.0","26.3","27.8","100.5","22.9","23.4"]))

def valid_readings(filepath):
    with open(filepath) as f:
        for line in f:
            val = float(line.strip())
            if 0 <= val <= 50:
                yield val

print(list(valid_readings("sensor_log.txt")))
```

## Q4 — Regex Extraction

```python
import re

text = """
Contact us: support@datagrokr.com or sales@datagrokr.com
Phone: +91-9876543210 or (080) 4567-8901
Invalid: not-an-email, 12345
"""
emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.-]+', text)
print("Emails:", emails)

phones_raw = re.findall(r'[\+\(]?[\d][\d\-\(\)\s]{8,}\d', text)
phones = [re.sub(r'\D', '', p) for p in phones_raw]
print("Phones (digits only):", phones)
print("Email count:", len(emails))
```

## Q5 — pandas: groupby & Aggregation

```python
import pandas as pd

sales = pd.DataFrame({
    "region": ["North","South","North","East","South","North","East"],
    "product": ["Widget","Widget","Gadget","Widget","Gadget","Gadget","Gadget"],
    "amount": [1200, 800, 1500, 900, 600, 1100, 700]
})

total_by_region = sales.groupby("region")["amount"].sum().sort_values(ascending=False)
print("Total by region:")
print(total_by_region)

avg_by_product = sales.groupby("product")["amount"].mean()
print()
print("Average by product:")
print(avg_by_product)

combo_totals = sales.groupby(["region","product"])["amount"].sum()
top_combo = combo_totals.idxmax()
top_value = combo_totals.max()
print()
print(f"Highest combo: {top_combo} with total {top_value}")
```

## Q6 — numpy: Descriptive Statistics

```python
import numpy as np

scores = np.array([67, 89, 45, 78, 92, 55, 88, 73, 61, 95])

print("Mean:", np.mean(scores))
print("Median:", np.median(scores))
print("Std Dev:", round(np.std(scores), 1))
print("90th percentile:", np.percentile(scores, 90))

labels = np.where(scores >= 60, "Pass", "Fail")
print("Labels:", labels)

above_mean = (scores > np.mean(scores)).sum()
print("Above mean count:", above_mean)
```

## Q7 — pytest: Writing Test Cases

## Q8 — SQL Query Writing

```python
import sqlite3

conn = sqlite3.connect(":memory:")
conn.execute("""
CREATE TABLE employees (
    id INTEGER PRIMARY KEY, name TEXT, dept TEXT, salary INTEGER, hire_date TEXT
)
""")
rows = [
    (1, "Asha", "Eng", 90000, "2021-03-15"),
    (2, "Ravi", "HR", 55000, "2022-06-01"),
    (3, "Zoya", "Eng", 95000, "2020-01-10"),
    (4, "Kabir", "Sales", 60000, "2023-02-20"),
    (5, "Meera", "Eng", 70000, "2022-09-05"),
]
conn.executemany("INSERT INTO employees VALUES (?,?,?,?,?)", rows)
conn.commit()

q1 = """
SELECT dept, AVG(salary) as avg_salary
FROM employees
GROUP BY dept
HAVING AVG(salary) > 60000
"""
print("Q1:", conn.execute(q1).fetchall())

q2 = """
SELECT name, hire_date FROM employees
ORDER BY hire_date DESC
LIMIT 2
"""
print("Q2:", conn.execute(q2).fetchall())

q3 = """
SELECT name FROM employees
WHERE dept = 'Eng' AND salary > (SELECT AVG(salary) FROM employees)
"""
print("Q3:", conn.execute(q3).fetchall())
```

**Note:** the original question's stated expected output had two mistakes, caught only by actually running the query — `Sales`'s average salary is *exactly* 60000, which fails `> 60000`, so it's correctly excluded from Q1; and Q2's second-most-recent hire is **Meera** (2022-09-05), not Ravi (2022-06-01).

## Q9 — Flask REST API Design

```python
from flask import Flask, jsonify

app = Flask(__name__)
students = {
    1: {"name": "Asha", "scores": [85, 90, 78]},
    2: {"name": "Ravi", "scores": [60, 55, 70]},
}

@app.route("/students/<int:sid>/grade")
def student_grade(sid):
    if sid not in students:
        return jsonify({"error": "student not found"}), 404
    s = students[sid]
    avg = round(sum(s["scores"]) / len(s["scores"]), 2)
    if avg >= 85: grade = "A"
    elif avg >= 70: grade = "B"
    elif avg >= 50: grade = "C"
    else: grade = "F"
    return jsonify({"name": s["name"], "average": avg, "grade": grade}), 200

client = app.test_client()
for sid in [1, 2, 99]:
    r = client.get(f"/students/{sid}/grade")
    print(sid, r.status_code, r.get_json())
```

## Q10 — End-to-End Mini Pipeline

```python
import re
import pandas as pd

lines = [
    "ORD001 | Customer: John Doe | Amount: $1,200.50 | Status: completed",
    "ORD002 | Customer: Jane Roe | Amount: $450 | Status: pending",
    "ORD003 | Customer: Sam Lee | Amount: $2,000.00 | Status: completed",
    "ORD004 | Customer: Amy Kim | Amount: $99.99 | Status: cancelled",
    "ORD005 | Customer: Leo Fox | Amount: $780.25 | Status: completed",
]
with open("raw_orders.txt", "w") as f:
    f.write("\n".join(lines))

rows = []
with open("raw_orders.txt") as f:
    for line in f:
        oid = re.search(r'(ORD\d+)', line).group(1)
        cust = re.search(r'Customer:\s*([^|]+)', line).group(1).strip()
        amt = float(re.search(r'Amount:\s*\$([\d,]+\.?\d*)', line).group(1).replace(",", ""))
        status = re.search(r'Status:\s*(\w+)', line).group(1)
        rows.append({"order_id": oid, "customer": cust, "amount": amt, "status": status})

df = pd.DataFrame(rows)
total_completed = df[df["status"] == "completed"]["amount"].sum()
print("Total completed amount:", total_completed)
print(df)
```
