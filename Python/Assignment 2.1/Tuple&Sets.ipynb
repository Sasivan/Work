```python
# ==========================================
# 1. NAMED TUPLES & GEOMETRY (Selected: Q89)
# ==========================================
# Question 89:
# Use collections.namedtuple to define a Point3D(x, y, z). Write functions: euclidean_distance(p1, p2), 
# midpoint(p1, p2), and translate(p, dx, dy, dz). Demonstrate with at least three points.
#
# Sample Input:  p1 = Point3D(1, 2, 3), p2 = Point3D(4, 6, 8), p3 = Point3D(-1, -2, -3)
# Sample Output: (7.071, Point3D(x=2.5, y=4.0, z=5.5), Point3D(x=3, y=5, z=7))

import math
def euclidean_distance(p1, p2):
    return math.hypot(
        p2.x-p1.x,
        p2.y-p1.y,
        p2.z-p1.z)

def midpoint(p1, p2):
    return Point3D(
        (p1.x+p2.x)/2,
        (p1.y+p2.y)/2,
        (p1.z+p2.z)/2
    )

def translate(p, dx, dy, dz):
    return Point3D(
        p.x+dx,
        p.y+dy,
        p.z+dz
    )

if __name__ == '__main__':
    # Test Question 89
    from collections import namedtuple
    Point3D = namedtuple('Point3D', ['x', 'y', 'z'])
    p1 = Point3D(1, 2, 3)
    p2 = Point3D(4, 6, 8)
    p3 = Point3D(-1, -2, -3)
    dist = round(euclidean_distance(p1, p2), 3) if euclidean_distance(p1, p2) is not None else None
    mid = midpoint(p1, p2)
    trans = translate(p1, 2, 3, 4)
    print("Q89 Output:", (dist, mid, trans))
```

```python
# ==========================================
# 2. SETS & RELATIONSHIPS (Selected: Q90)
# ==========================================
# Question 90:
# Write a function that takes a list of (student, course) tuples and returns a dictionary where each 
# student maps to a frozenset of courses they are enrolled in. Then find all pairs of students who share at 
# least two common courses.
#
# Sample Input:  enrollments = [("Alice", "Math"), ("Alice", "CS"), ("Alice", "Physics"), ("Bob", "Math"), ("Bob", "CS"), ("Charlie", "Math")]
# Sample Output: ({'Alice': frozenset({'Math', 'CS', 'Physics'}), 'Bob': frozenset({'Math', 'CS'}), 'Charlie': frozenset({'Math'})}, [(('Alice', 'Bob'), frozenset({'Math', 'CS'}))])

def map_students_to_courses(enrollments):
    res = {}
    for i in enrollments:
        if i[0] not in res:
            res[i[0]] = {i[1]}
        else: res[i[0]].add(i[1])
    for k,v in res.items():
        res[k] = frozenset(v)
    return res

def find_shared_course_pairs(student_courses, min_common=2):
    students = list(student_courses)
    res = []

    for i in range(len(students)):
        for j in range(i + 1, len(students)):
            s1 = students[i]
            s2 = students[j]

            common = student_courses[s1] & student_courses[s2]

            if len(common) >= min_common:
                res.append(((s1, s2), common))

    return res

if __name__ == '__main__':
    # Test Question 90
    enrollments = [("Alice", "Math"), ("Alice", "CS"), ("Alice", "Physics"), ("Bob", "Math"), ("Bob", "CS"), ("Charlie", "Math")]
    mapped = map_students_to_courses(enrollments)
    pairs = find_shared_course_pairs(mapped, min_common=2)
    print("Q90 Output:", (mapped, pairs))
```

```python
# ==========================================
# 3. INVERTED INDEX & SET OPERATIONS (Selected: Q91)
# ==========================================
# Question 91:
# Write a program that reads a list of (word, document_id) tuples representing an inverted index, builds 
# a set-based inverted index dictionary, and supports AND, OR, and NOT queries (e.g., "python AND tutorial 
# NOT beginner").
#
# Sample Input:  index_data = [("python", 1), ("python", 2), ("tutorial", 2), ("tutorial", 3), ("beginner", 1)]
# Sample Output: {2}

def build_inverted_index(data):
    res = {}
    for i in data:
        if i[0] not in res:
            res[i[0]] = set()
        res[i[0]].add(i[1])
    return res

def execute_query(index, query_str, all_doc_ids):
    query = query_str.split(" ")
    res = set()
    op = None
    for i in query:
        if i in ["AND","OR","NOT"]:
            op = i
        else:
            doc = set(index[i])
            if not res: res = doc
            elif op == "AND": res &= doc
            elif op == "OR": res |= doc
            else: res -= doc
    return res

if __name__ == '__main__':
    # Test Question 91
    index_data = [("python", 1), ("python", 2), ("tutorial", 2), ("tutorial", 3), ("beginner", 1)]
    idx = build_inverted_index(index_data)
    all_ids = {1, 2, 3}
    print("Q91 Output:", execute_query(idx, "python AND tutorial NOT beginner", all_ids))
```

```python
# ==========================================
# 4. POWER SETS & FROZENSETS (Selected: Q92)
# ==========================================
# Question 92:
# Write a function power_set(s) that returns the power set of a given set as a set of frozensets. For 
# example, {1,2,3} → frozensets of: {}, {1}, {2}, {3}, {1,2}, {1,3}, {2,3}, {1,2,3}.
#
# Sample Input:  s = {1, 2, 3}
# Sample Output: {frozenset(), frozenset({1}), frozenset({2}), frozenset({3}), frozenset({1, 2}), frozenset({1, 3}), frozenset({2, 3}), frozenset({1, 2, 3})}

def power_set(s):
    res = set()

    def func(i,curr):
        if i == len(s):
            res.add(frozenset(curr))
            return
        func(i+1,curr)

        curr.add(s[i])
        func(i+1,curr)
        curr.remove(s[i])   

    s = list(s)
    func(0,set())
    return res

if __name__ == '__main__':
    # Test Question 92
    print("Q92 Output:", power_set({1, 2, 3}))
```

```python
# ==========================================
# 5. SET COVERAGE & INTERSECTIONS (Selected: Q93)
# ==========================================
# Question 93:
# Given a list of sets, write a function that finds the set(s) with the maximum number of elements in 
# common with all other sets combined. Print the winning set and its union-coverage percentage.
#
# Sample Input:  sets_list = [{1, 2, 3, 4}, {3, 4, 5, 6}, {4, 5, 6, 7}]
# Sample Output: ({3, 4, 5, 6}, 57.14)

def find_best_coverage_set(sets_list):
    res = []
    u = set().union(*sets_list)

    for i in sets_list:
        res.append((i,round(100*len(i)/len(u),2)))
    return max(res,key=lambda x:x[1])
if __name__ == '__main__':
    # Test Question 93
    sample_sets = [{1, 2, 3, 4}, {3, 4, 5, 6}, {4, 5, 6, 7}]
    print("Q93 Output:", find_best_coverage_set(sample_sets))
```

```python
# ==========================================
# 6. TUPLE DEDUPLICATION (Selected: Q94)
# ==========================================
# Question 94:
# Write a function that takes a tuple of integers and returns a new tuple where consecutive duplicates 
# are collapsed into one occurrence. For example, (1,1,2,3,3,3,4,4) → (1,2,3,4). Do not use itertools.
#
# Sample Input:  tup = (1, 1, 2, 3, 3, 3, 4, 4)
# Sample Output: (1, 2, 3, 4)

def collapse_consecutive_duplicates(tup):
    res = []

    for i in tup:
        if len(res) > 0 and res[-1] == i: continue
        res.append(i)
    return tuple(res)
if __name__ == '__main__':
    # Test Question 94
    print("Q94 Output:", collapse_consecutive_duplicates((1, 1, 2, 3, 3, 3, 4, 4)))
```
