```python
# ==========================================
# 14. TESTING & DEBUGGING (Selected: Q148)
# ==========================================
# Question 148:
# Practice TDD: write the test class first for a RomanConverter class (integer <-> Roman, 1–3999), then 
# implement the class to make all tests pass. Do not touch the tests once written. Tests must cover at least: 
# boundary values (1, 3999), all subtractive pairs (IV, IX, XL, etc.), and invalid inputs.
#
# Sample Input:  converter = RomanConverter(); converter.to_roman(1994)
# Sample Output: 'MCMXCIV' 

import unittest


class RomanConverter:

    def __init__(self):
        self.val_map = [
            (1000, 'M'),
            (900, 'CM'),
            (500, 'D'),
            (400, 'CD'),
            (100, 'C'),
            (90, 'XC'),
            (50, 'L'),
            (40, 'XL'),
            (10, 'X'),
            (9, 'IX'),
            (5, 'V'),
            (4, 'IV'),
            (1, 'I')
        ]

    def to_roman(self, num):
        if not isinstance(num, int) or num < 1 or num > 3999:
            raise ValueError("Number must be between 1 and 3999")

        result = ""

        for value, symbol in self.val_map:
            while num >= value:
                result += symbol
                num -= value

        return result

    def from_roman(self, roman):
        if not isinstance(roman, str) or not roman:
            raise ValueError("Invalid Roman numeral")

        values = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }

        total = 0

        for i in range(len(roman)):
            if roman[i] not in values:
                raise ValueError("Invalid Roman numeral")

            if i + 1 < len(roman) and values[roman[i]] < values[roman[i + 1]]:
                total -= values[roman[i]]
            else:
                total += values[roman[i]]

        if total < 1 or total > 3999:
            raise ValueError("Invalid Roman numeral")

        return total


class TestRomanConverter(unittest.TestCase):

    def setUp(self):
        self.converter = RomanConverter()

    def test_boundaries(self):
        self.assertEqual(self.converter.to_roman(1), "I")
        self.assertEqual(self.converter.to_roman(3999), "MMMCMXCIX")

        self.assertEqual(self.converter.from_roman("I"), 1)
        self.assertEqual(self.converter.from_roman("MMMCMXCIX"), 3999)

    def test_subtractive_pairs(self):
        self.assertEqual(self.converter.to_roman(4), "IV")
        self.assertEqual(self.converter.to_roman(9), "IX")
        self.assertEqual(self.converter.to_roman(40), "XL")
        self.assertEqual(self.converter.to_roman(90), "XC")
        self.assertEqual(self.converter.to_roman(400), "CD")
        self.assertEqual(self.converter.to_roman(900), "CM")

        self.assertEqual(self.converter.from_roman("IV"), 4)
        self.assertEqual(self.converter.from_roman("IX"), 9)
        self.assertEqual(self.converter.from_roman("XL"), 40)
        self.assertEqual(self.converter.from_roman("XC"), 90)
        self.assertEqual(self.converter.from_roman("CD"), 400)
        self.assertEqual(self.converter.from_roman("CM"), 900)

    def test_invalid_inputs(self):
        with self.assertRaises(ValueError):
            self.converter.to_roman(0)

        with self.assertRaises(ValueError):
            self.converter.to_roman(4000)

        with self.assertRaises(ValueError):
            self.converter.to_roman("100")

        with self.assertRaises(ValueError):
            self.converter.from_roman("ABC")

        with self.assertRaises(ValueError):
            self.converter.from_roman("")

        with self.assertRaises(ValueError):
            self.converter.from_roman(123)


if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

```python
# ==========================================
# 14. TESTING & DEBUGGING (Selected: Q150)
# ==========================================
# Question 150:
# Write a performance benchmarking harness using timeit and the statistics module. It should: run a 
# given callable n times, collect timings, and print min, max, mean, median, and standard deviation. Use it to 
# compare your merge sort (Q65) against Python's built-in sorted() on lists of size 100, 1000, and 10000.
#
# Sample Input:  benchmark(sorted, sample_data, runs=10)
# Sample Output: {'min': 0.0001, 'max': 0.0003, 'mean': 0.00015, 'median': 0.00014, 'stdev': 0.00002}

import timeit
import statistics


def custom_merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = custom_merge_sort(arr[:mid])
    right = custom_merge_sort(arr[mid:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


def benchmark(func, data_generator, runs=10):

    times = []

    for _ in range(runs):

        data = data_generator()

        start = timeit.default_timer()

        func(data)

        end = timeit.default_timer()

        times.append(end - start)

    result = {
        "min": min(times),
        "max": max(times),
        "mean": statistics.mean(times),
        "median": statistics.median(times),
        "stdev": statistics.stdev(times) if runs > 1 else 0
    }

    print(result)

    return result


if __name__ == '__main__':

    print(
        "Q150 Output:",
        benchmark(
            sorted,
            lambda: [10, 5, 2, 8, 1],
            runs=5
        )
    )

import random

for size in [100, 1000, 10000]:

    print("Size:", size)

    data_generator = lambda size=size: [
        random.randint(1, 100000)
        for _ in range(size)
    ]

    print("Merge Sort:")
    benchmark(custom_merge_sort, data_generator, 10)

    print("Python sorted:")
    benchmark(sorted, data_generator, 10)
```
