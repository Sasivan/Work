```python
# ==========================================
# 1. DICTIONARY MANIPULATION (Selected: Q79)
# ==========================================
# Question 79:
# Write a function flatten_nested_dict(d, sep=".") that flattens a deeply nested dictionary into a single
# level dictionary where nested keys are joined by a separator. For example, {"a": {"b": {"c": 1}}} becomes 
# {"a.b.c": 1}. Handle mixed nesting depths.
#
# Sample Input:  d = {"a": 1, "b": {"c": 2, "d": {"e": 3}}, "f": {"g": 4}}
# Sample Output: {'a': 1, 'b.c': 2, 'b.d.e': 3, 'f.g': 4}

def flatten_nested_dict(d, sep="."):
    res = {}
    def flatten(data,prefix=""):
        for k,v in data.items():
            key = prefix+sep+k if prefix else k
            if isinstance(v,dict):
                flatten(v,key) 
            else: res[key] = v
    flatten(d)
    return res

if __name__ == '__main__':
    # Test Question 79
    sample_dict = {"a": 1, "b": {"c": 2, "d": {"e": 3}}, "f": {"g": 4}}
    print("Q79 Output:", flatten_nested_dict(sample_dict))
```

```python
#DFS & BFS

from collections import deque

def bfs(graph,start):
    visited = set()
    queue = deque({start})
    res = []

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.add(node)
            res.append(node)

            for neighbour in graph[node]:
                queue.append(neighbour)
    return res

res = []
def dfs(graph,start,vis = None):
    global res
    if vis is None: vis = set()
    vis.add(start)
    res.append(start)
    for neighbour in graph[start]:
        if neighbour not in vis:
            dfs(graph,neighbour,vis)
    return res


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

start = 'A'
print(bfs(graph, 'A'))
print(dfs(graph, 'A'))
```

```python
# ==========================================
# 2. GRAPH DATA STRUCTURES (Selected: Q80)
# ==========================================
# Question 80:
# Represent a directed graph as a dictionary of adjacency lists. Write functions: add_edge(graph, u, v), 
# remove_edge(graph, u, v), has_path(graph, start, end) using BFS, and find_all_paths(graph, start, end) 
# using DFS. Demonstrate on a sample graph.
#
# Sample Input:  graph = {'A': ['B', 'C'], 'B': ['D'], 'C': ['D'], 'D': ['E'], 'E': []}
# Sample Output: (True, [['A', 'B', 'D', 'E'], ['A', 'C', 'D', 'E']])

def add_edge(graph, u, v):
    if u not in graph: graph[u] = []
    graph[u].append(v)

def remove_edge(graph, u, v):
    if u in graph and v in graph[u]:
        graph[u].remove(v)

def has_path(graph, start, end):
    vis = set()
    queue = deque([start])
    # res = [start]
    while queue:
        node = queue.popleft()
        if node == end: 
            # print(res)
            return True
        if node in vis: continue
        vis.add(node)
        for neighbour in graph[node]:
            queue.append(neighbour)
            # res.append(neighbour)
    return False

def find_all_paths(graph, start, end):
    paths = []
    def dfs(node,path):
        if node == end:
            paths.append(path.copy())
            return
        for neighbour in graph[node]:
            if neighbour not in path:
                path.append(neighbour)
                dfs(neighbour,path)
                path.pop()
    dfs(start,[start])
    return paths

if __name__ == '__main__':
    # Test Question 80
    graph = {'A': ['B', 'C'], 'B': ['D'], 'C': ['D'], 'D': ['E'], 'E': []}
    has_p = has_path(graph, 'A', 'E')
    all_p = find_all_paths(graph, 'A', 'E')
    print("Q80 Output:", (has_p, all_p))
```

```python
# ==========================================
# 3. FREQUENCY MAPS & SORTING (Selected: Q81)
# ==========================================
# Question 81:
# Write a function top_k_frequent(words, k) that takes a list of words and returns the k most frequently 
# occurring words, sorted by frequency descending and alphabetically for ties. Do not use Counter — build 
# the frequency map manually.
#
# Sample Input:  words = ["i", "love", "leetcode", "i", "love", "coding"], k = 2
# Sample Output: ['i', 'love']

def top_k_frequent(words, K):
    res = {}
    for k in words:
        if k not in res:
            res[k] = words.count(k)
    out = sorted(res,key=lambda x:(-res[x],x))[:K]
    return out

if __name__ == '__main__':
    # Test Question 81
    print("Q81 Output:", top_k_frequent(["i", "love", "leetcode", "i", "love", "coding"], 2))
```

```python
# ==========================================
# 4. LOG PROCESSING & GROUPING (Selected: Q82)
# ==========================================
# Question 82:
# Write a function that takes a list of log entries (strings in the format "YYYY-MM-DD HH:MM:SS LEVEL 
# message") and groups them into a nested dictionary: {date: {level: [messages]}}. Print the summary: total 
# messages per date and per level.
#
# Sample Input:  logs = ["2026-08-10 10:00:00 INFO User logged in", "2026-08-10 10:05:00 ERROR Connection failed", "2026-08-11 11:00:00 INFO System started"]
# Sample Output: ({'2026-08-10': {'INFO': ['User logged in'], 'ERROR': ['Connection failed']}, '2026-08-11': {'INFO': ['System started']}}, {'2026-08-10': {'INFO': 1, 'ERROR': 1}, '2026-08-11': {'INFO': 1}})

def parse_and_group_logs(logs):
    res = {}
    summary = {}
    for i in logs:
        date = i.split(' ')[0]
        level = i.split(' ')[2]
        msg = ' '.join(i.split(' ')[3:])

        if date not in res: 
            res[date] = {}
            summary[date] = {}
        if level not in res[date]: 
            res[date][level] = []
            summary[date][level] = 1
        else: summary[level] += 1
        res[date][level].append(msg)
        
    return res,summary

        
if __name__ == '__main__':
    # Test Question 82
    sample_logs = [
        "2026-08-10 10:00:00 INFO User logged in",
        "2026-08-10 10:05:00 ERROR Connection failed",
        "2026-08-11 11:00:00 INFO System started"
    ]
    print("Q82 Output:", parse_and_group_logs(sample_logs))
```

```python
# ==========================================
# 5. CACHING DATA STRUCTURES (Selected: Q83)
# ==========================================
# Question 83:
# Implement an LRU (Least Recently Used) cache using only a dictionary and a list. It should support 
# get(key) → value or -1, and put(key, value). When capacity is exceeded, evict the least recently used entry. 
# Demonstrate with at least 8 operations.
#
# Sample Input:  capacity = 2, operations = [put(1, 1), put(2, 2), get(1), put(3, 3), get(2), put(4, 4), get(1), get(3)]
# Sample Output: [1, -1, -1, 3]

from collections import deque
class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.data = {}
        self.store = deque()
    def get(self, key):
        if key not in self.data: return -1
        else:
            self.store.remove(key)
            self.store.append(key)
            return self.data[key]

    def put(self, key, value):
        if key in self.store: 
            self.store.remove(key)
        elif len(self.store) == self.capacity:
            k = self.store.popleft()
            del self.data[k]
        self.store.append(key)
        self.data[key] = value

if __name__ == '__main__':
    # Test Question 83
    cache = LRUCache(2)
    cache.put(1, 1)
    cache.put(2, 2)
    res1 = cache.get(1)
    cache.put(3, 3)
    res2 = cache.get(2)
    cache.put(4, 4)
    res3 = cache.get(1)
    res4 = cache.get(3)
    print("Q83 Output:", [res1, res2, res3, res4])
```

```python
# ==========================================
# 6. DICTIONARY RECURSION & DIFFING (Selected: Q84)
# ==========================================
# Question 84:
# Write a function diff_dicts(old, new_dict) that returns a dictionary describing changes: {"added": {k:v}, 
# "removed": {k:v}, "modified": {k: {"old": v, "new": v}}}. Handle nested dictionaries recursively.
#
# Sample Input:  old = {"a": 1, "b": 2, "c": {"d": 3}}, new_dict = {"b": 20, "c": {"d": 3, "e": 4}, "f": 5}
# Sample Output: {'added': {'f': 5, 'c': {'added': {'e': 4}}}, 'removed': {'a': 1}, 'modified': {'b': {'old': 2, 'new': 20}}}

def diff_dicts(old, new_dict):
    res = {"added":{},"removed":{},"modified":{}}
    keys = old.keys() | new_dict.keys()

    for key in keys:
        if key not in old:
            res["added"][key] = new_dict[key]
        elif key not in new_dict:
            res["removed"][key] = old[key]
        elif old[key] != new_dict[key]:
            if isinstance(old[key],dict) and isinstance(new_dict[key],dict):
                diff = diff_dicts(old[key],new_dict[key])
                # print(diff)
                if diff["added"]:
                    res["added"][key] = {"added":diff["added"]}

            else:
                res["modified"][key] = {
                    "old":old[key],
                    "new":new_dict[key]
                }

    return res

if __name__ == '__main__':
    # Test Question 84
    d1 = {"a": 1, "b": 2, "c": {"d": 3}}
    d2 = {"b": 20, "c": {"d": 3, "e": 4}, "f": 5}
    print("Q84 Output:", diff_dicts(d1, d2))
```

```python
# ==========================================
# 7. ANAGRAM GROUPING (Selected: Q85)
# ==========================================
# Question 85:
# Write a function group_anagrams(words) that groups a list of words into clusters of anagrams. Return 
# a list of lists. For example, ["eat","tea","tan","ate","nat","bat"] becomes 
# [["eat","tea","ate"],["tan","nat"],["bat"]]. Use only dictionaries — no sorting of words.
#
# Sample Input:  words = ["eat", "tea", "tan", "ate", "nat", "bat"]
# Sample Output: [['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]

def group_anagrams(words):
    res = {}
    for word in words:
        count = [0]*26
        for ch in word:
            count[ord(ch) - ord('a')] += 1
        key =  tuple(count)
        if key not in res:
            res[key] = []
        res[key].append(word)
        # print(res)
    return list(res.values())

if __name__ == '__main__':
    # Test Question 85
    print("Q85 Output:", group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

```python
# ==========================================
# 8. TREE / JSON TRAVERSAL (Selected: Q86)
# ==========================================
# Question 86:
# Write a program that reads a JSON string representing a company org chart (employees with name, 
# title, and reports: [...]) and produces: (a) a flattened list of all employees, (b) the depth of the deepest 
# branch, (c) a count of employees per level.
#
# Sample Input:  json_str = '{"name": "Alice", "title": "CEO", "reports": [{"name": "Bob", "title": "VP", "reports": []}, {"name": "Charlie", "title": "VP", "reports": [{"name": "David", "title": "Manager", "reports": []}]}]}'
# Sample Output: (['Alice', 'Bob', 'Charlie', 'David'], 3, {1: 1, 2: 2, 3: 1})

import json
def parse_org_chart(json_str):
    data = json.loads(json_str)
    emp = []
    levels = {}

    def dfs(node,level):
        emp.append(node["name"])
        levels[level] = levels.get(level,0)+1
        for p in node["reports"]:
            dfs(p,level+1)
    dfs(data,1)
    return emp,max(levels),levels


if __name__ == '__main__':
    # Test Question 86
    sample_json = '{"name": "Alice", "title": "CEO", "reports": [{"name": "Bob", "title": "VP", "reports": []}, {"name": "Charlie", "title": "VP", "reports": [{"name": "David", "title": "Manager", "reports": []}]}]}'
    print("Q86 Output:", parse_org_chart(sample_json))
```

```python
# ==========================================
# 9. FINANCIAL DATA PROCESSING (Selected: Q87)
# ==========================================
# Question 87:
# Write a function that takes a list of transactions, each a dict with keys date, account, type 
# ("credit"/"debit"), and amount, and produces a per-account running balance history as a dictionary of 
# {account: [(date, balance_after_transaction)]}.
#
# Sample Input:  transactions = [{'date': '2026-08-01', 'account': 'A1', 'type': 'credit', 'amount': 100}, {'date': '2026-08-02', 'account': 'A1', 'type': 'debit', 'amount': 30}, {'date': '2026-08-01', 'account': 'A2', 'type': 'credit', 'amount': 200}]
# Sample Output: {'A1': [('2026-08-01', 100), ('2026-08-02', 70)], 'A2': [('2026-08-01', 200)]}

def calculate_running_balances(transactions):
    res = {}
    balance = {}

    for i in transactions:
        account = i["account"]
        if account not in res:
            res[account] = []
            balance[account] = 0
        if i["type"] == "credit":
            balance[account] += i["amount"]
        else:
            balance[account] -= i["amount"]

        res[account].append(
            (i["date"], balance[account])
        )

    return res,balance

if __name__ == '__main__':
    # Test Question 87
    sample_txs = [
        {'date': '2026-08-01', 'account': 'A1', 'type': 'credit', 'amount': 100},
        {'date': '2026-08-02', 'account': 'A1', 'type': 'debit', 'amount': 30},
        {'date': '2026-08-01', 'account': 'A2', 'type': 'credit', 'amount': 200}
    ]
    print("Q87 Output:", calculate_running_balances(sample_txs))
```

```python
# ==========================================
# 10. BREADTH-FIRST SEARCH & GRAPH SHORTEST PATH (Selected: Q88)
# ==========================================
# Question 88:
# Write a function word_ladder_distance(start, end, word_list) that finds the minimum number of one
# letter changes needed to transform start into end (each intermediate word must be in word_list). Use BFS. 
# A dictionary will be helpful for O(1) lookups.
#
# Sample Input:  start = "hit", end = "cog", word_list = ["hot", "dot", "dog", "lot", "log", "cog"]
# Sample Output: 5

from collections import deque
def word_ladder_distance(start, end, word_list):
    words = set(word_list)
    queue = deque([(start,1)])
    vis = {start}

    while queue:
        word,dis = queue.popleft()

        if word == end: return dis

        for i in range(len(word)):
            for ch in "abcdefghijklmnopqrstuvwxyz":
                new_word = word[:i] + ch + word[i+1:]

                if new_word in words and new_word not in vis:
                    vis.add(new_word)
                    queue.append((new_word,dis+1))
    return 0
if __name__ == '__main__':
    # Test Question 88
    words = ["hot", "dot", "dog", "lot", "log", "cog"]
    print("Q88 Output:", word_ladder_distance("hit", "cog", words))
```
