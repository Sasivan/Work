```python
import abc
import math
import threading
from typing import Any, Callable, Dict, List

# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q130)
# ==========================================
# Question 130:
# Write an abstract base class Animal using abc.ABC with abstract methods speak(), move(), and eat(). 
# Create four concrete subclasses: Dog, Cat, Bird, Fish — each implementing all three methods with realistic 
# behaviour. Write a function animal_report(animals: list[Animal]) that polymorphically calls all three 
# methods on each animal.
#
# Sample Input:  animals = [Dog(), Cat(), Bird(), Fish()]
# Sample Output: [('Dog: Woof!', 'Dog runs on 4 legs.', 'Dog eats meat.'), ('Cat: Meow!', 'Cat walks stealthily.', 'Cat eats fish.'), ('Bird: Chirp!', 'Bird flies in the sky.', 'Bird eats seeds.'), ('Fish: Blub!', 'Fish swims in water.', 'Fish eats algae.')]

class Animal(abc.ABC):
    @abc.abstractmethod
    def speak(self) -> str:
        pass

    @abc.abstractmethod
    def move(self) -> str:
        pass

    @abc.abstractmethod
    def eat(self) -> str:
        pass

class Dog(Animal):
    def speak(self) -> str:
        return "Dog: Woof!"

    def move(self) -> str:
        return "Dog runs on 4 legs."

    def eat(self) -> str:
        return "Dog eats meat."

class Cat(Animal):
    def speak(self) -> str:
        return "Cat: Meow!"

    def move(self) -> str:
        return "Cat walks stealthily."

    def eat(self) -> str:
        return "Cat eats fish."

class Bird(Animal):
    def speak(self) -> str:
        return "Bird: Chirp!"

    def move(self) -> str:
        return "Bird flies in the sky."

    def eat(self) -> str:
        return "Bird eats seeds."

class Fish(Animal):
    def speak(self) -> str:
        return "Fish: Blub!"

    def move(self) -> str:
        return "Fish swims in water."

    def eat(self) -> str:
        return "Fish eats algae."

def animal_report(animals: List[Animal]):
    return [(animal.speak(), animal.move(), animal.eat()) for animal in animals]

if __name__ == '__main__':
    # Test Question 130
    pass_animals = [Dog(), Cat(), Bird(), Fish()]
    print("Q130 Output:", animal_report(pass_animals))
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q131)
# ==========================================
# Question 131:
# Implement a generic Stack class and a Queue class. Both should support push/enqueue, 
# pop/dequeue, peek, size, is_empty, and __len__. Override __repr__ and __str__. Implement __iter__ so 
# both are iterable (non-destructively).
#
# Sample Input:  s = Stack(); s.push(1); s.push(2); q = Queue(); q.enqueue(10); q.enqueue(20)
# Sample Output: {'stack_pop': 2, 'stack_iter': [1], 'queue_dequeue': 10, 'queue_iter': [20]}

class Stack:
    def __init__(self): self._data = []
    def push(self, x): self._data.append(x)
    def pop(self): return self._data.pop()
    def peek(self): return self._data[-1]
    def is_empty(self): return len(self._data) == 0
    def size(self): return len(self._data)
    def __len__(self): return len(self._data)
    def __iter__(self): return iter(self._data)
    def __repr__(self): return f"Stack({self._data})"

class Queue:
    def __init__(self): self._data = []
    def enqueue(self, x): self._data.append(x)
    def dequeue(self): return self._data.pop(0)
    def peek(self): return self._data[0]
    def is_empty(self): return len(self._data) == 0
    def size(self): return len(self._data)
    def __len__(self): return len(self._data)
    def __iter__(self): return iter(self._data)
    def __repr__(self): return f"Queue({self._data})"

if __name__ == '__main__':
    s, q = Stack(), Queue()
    s.push(1); s.push(2)
    q.enqueue(10); q.enqueue(20)
    print("Q131 Output:", {
        'stack_pop': s.pop(), 'stack_iter': list(s),
        'queue_dequeue': q.dequeue(), 'queue_iter': list(q)
    })
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q132)
# ==========================================
# Question 132:
# Write a class Vector that represents a mathematical vector of arbitrary dimension. Overload: +, -, * 
# (scalar and dot product), == , len (dimensionality), and abs (magnitude). Implement __repr__ and verify all 
# operations with unit tests using assert.
#
# Sample Input:  v1 = Vector(1, 2, 3), v2 = Vector(4, 5, 6)
# Sample Output: {'add': Vector(5.0, 7.0, 9.0), 'sub': Vector(-3.0, -3.0, -3.0), 'dot': 32.0, 'scalar_mul': Vector(2.0, 4.0, 6.0), 'mag': 3.742}

class Vector:
    def __init__(self, *components):
        self.c = tuple(float(x) for x in components)

    def __add__(self, o): return Vector(*(a + b for a, b in zip(self.c, o.c)))
    def __sub__(self, o): return Vector(*(a - b for a, b in zip(self.c, o.c)))
    def __mul__(self, o):
        if isinstance(o, (int, float)):
            return Vector(*(a * o for a in self.c))
        return sum(a * b for a, b in zip(self.c, o.c))
    def __rmul__(self, o): return self * o
    def __eq__(self, o): return isinstance(o, Vector) and self.c == o.c
    def __len__(self): return len(self.c)
    def __abs__(self): return math.sqrt(sum(x**2 for x in self.c))
    def __repr__(self): return f"Vector{self.c}"

if __name__ == '__main__':
    v1, v2 = Vector(1, 2, 3), Vector(4, 5, 6)
    print("Q132 Output:", {
        'add': v1 + v2, 'sub': v1 - v2, 'dot': v1 * v2,
        'scalar_mul': v1 * 2, 'mag': round(abs(v1), 3)
    })
    
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q133)
# ==========================================
# Question 133:
# Implement a Singleton pattern using a metaclass. Write a Logger class that is a Singleton — calling 
# Logger() always returns the same instance. Verify this in at least three different ways. Then implement a 
# thread-safe version using threading.Lock.
#
# Sample Input:  l1 = ThreadSafeLogger(); l2 = ThreadSafeLogger()
# Sample Output: (True, True)

class SingletonMeta(type):
    _instances = {}
    _lock = threading.Lock()
    def __call__(cls, *args, **kwargs):
        with cls._lock:
            if cls not in cls._instances:
                cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Logger(metaclass=SingletonMeta):
    def __init__(self): self.logs = []

class ThreadSafeLogger(metaclass=SingletonMeta):
    def __init__(self): self.logs = []

if __name__ == '__main__':
    l1, l2 = ThreadSafeLogger(), ThreadSafeLogger()
    print("Q133 Output:", (l1 is l2, id(l1) == id(l2)))
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q134)
# ==========================================
# Question 134:
# Write a class Matrix with __add__, __sub__, __mul__ (scalar and matrix multiplication), __eq__, 
# __getitem__, __setitem__, __len__, and __repr__. Validate dimensions on operations and raise descriptive 
# errors. Test with a 3×3 × 3×3 multiplication.
#
# Sample Input:  m1 = Matrix([[1, 2], [3, 4]]), m2 = Matrix([[2, 0], [1, 2]])
# Sample Output: Matrix([[4, 4], [10, 8]])

class Matrix:
    def __init__(self, grid):
        self.g = [list(r) for r in grid]
        self.r, self.c = len(grid), len(grid[0])

    def __add__(self, o):
        return Matrix([[self.g[i][j] + o.g[i][j] for j in range(self.c)] for i in range(self.r)])

    def __sub__(self, o):
        return Matrix([[self.g[i][j] - o.g[i][j] for j in range(self.c)] for i in range(self.r)])

    def __mul__(self, o):
        if isinstance(o, (int, float)):
            return Matrix([[x * o for x in row] for row in self.g])
        return Matrix([[sum(self.g[i][k] * o.g[k][j] for k in range(self.c)) for j in range(o.c)] for i in range(self.r)])

    def __getitem__(self, idx): return self.g[idx]
    def __setitem__(self, idx, val): self.g[idx] = val
    def __len__(self): return self.r
    def __repr__(self): return f"Matrix({self.g})"

if __name__ == '__main__':
    m1 = Matrix([[1, 2], [3, 4]])
    m2 = Matrix([[2, 0], [1, 2]])
    print("Q134 Output:", m1 * m2)
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q135)
# ==========================================
# Question 135:
# Write a descriptor class TypeChecked that can be used as a class attribute to enforce a specific type 
# on assignment. For example, assigning a string to an attribute declared as TypeChecked(int) should raise 
# TypeError. Demonstrate with a Product class (name: str, price: float, quantity: int).
#
# Sample Input:  p = Product("Laptop", 999.99, 5)
# Sample Output: {'name': 'Laptop', 'price': 999.99, 'quantity': 5}

class TypeChecked:
    def __init__(self, typ): self.typ = typ
    def __set_name__(self, owner, name): self.name = f"_{name}"
    def __get__(self, inst, owner): return getattr(inst, self.name, None)
    def __set__(self, inst, val):
        if not isinstance(val, self.typ):
            raise TypeError(f"Expected {self.typ.__name__}, got {type(val).__name__}")
        setattr(inst, self.name, val)

class Product:
    name = TypeChecked(str)
    price = TypeChecked(float)
    quantity = TypeChecked(int)

    def __init__(self, name, price, quantity):
        self.name, self.price, self.quantity = name, price, quantity

if __name__ == '__main__':
    p = Product("Laptop", 999.99, 5)
    print("Q135 Output:", {'name': p.name, 'price': p.price, 'quantity': p.quantity})
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q136)
# ==========================================
# Question 136:
# Implement the Observer pattern: a class EventBus where objects can subscribe to named events and 
# receive callbacks when those events are published. Support: subscribe(event, handler), unsubscribe(event, 
# handler), publish(event, data), and once(event, handler) (fires only once). Demonstrate with three different 
# event types.
#
# Sample Input:  bus = EventBus(); bus.subscribe("login", on_login); bus.publish("login", {"user": "Alice"})
# Sample Output: ['Received login for Alice']

class EventBus:
    def __init__(self):
        self.subs = {}
        self.once_subs = {}

    def subscribe(self, event, fn): self.subs.setdefault(event, []).append(fn)
    def once(self, event, fn): self.once_subs.setdefault(event, []).append(fn)
    def unsubscribe(self, event, fn):
        if event in self.subs and fn in self.subs[event]: self.subs[event].remove(fn)

    def publish(self, event, data):
        for fn in list(self.subs.get(event, [])): fn(data)
        for fn in self.once_subs.pop(event, []): fn(data)

if __name__ == '__main__':
    logs = []
    bus = EventBus()
    bus.subscribe("user_login", lambda d: logs.append(f"Received login for {d.get('user')}"))
    bus.publish("user_login", {"user": "Alice"})
    print("Q136 Output:", logs)
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q137)
# ==========================================
# Question 137:
# Write a class Pipeline (the Builder pattern) for constructing data transformation pipelines. Each step 
# is a callable. Support: add_step(name, func), remove_step(name), run(data), and __repr__ showing the 
# ordered steps. Demonstrate with a text-processing pipeline of 5 steps.
#
# Sample Input:  pipeline.run("  HELLO WORLD  ")
# Sample Output: 'hello python'

class Pipeline:
    def __init__(self): self.steps = []
    def add_step(self, name, fn):
        self.steps.append((name, fn))
        return self
    def remove_step(self, name):
        self.steps = [s for s in self.steps if s[0] != name]
        return self
    def run(self, data):
        for _, fn in self.steps: data = fn(data)
        return data
    def __repr__(self): return f"Pipeline({' -> '.join(n for n, _ in self.steps)})"

if __name__ == '__main__':
    pipe = Pipeline().add_step("strip", str.strip).add_step("lower", str.lower).add_step("rep", lambda s: s.replace("world", "python"))
    print("Q137 Output:", pipe.run("  HELLO WORLD  "))
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q138)
# ==========================================
# Question 138:
# Design a class hierarchy for a simple e-commerce system: Product (name, price, stock), 
# DigitalProduct (download_url), PhysicalProduct (weight_kg, shipping_cost()), Cart (add, remove, total, 
# discount_code support), and Order (snapshot of cart, status tracking). Use __slots__ on Product to reduce 
# memory overhead.
#
# Sample Input:  cart = Cart(); cart.add(PhysicalProduct("Book", 20.0, 10, 0.5)); cart.apply_discount(0.1)
# Sample Output: {'total': 23.0, 'order_status': 'Placed'}

class ProductBase:
    __slots__ = ('name', 'price', 'stock')
    def __init__(self, name, price, stock):
        self.name, self.price, self.stock = name, float(price), int(stock)

class DigitalProduct(ProductBase):
    __slots__ = ('download_url',)
    def __init__(self, name, price, stock, download_url):
        super().__init__(name, price, stock)
        self.download_url = download_url

class PhysicalProduct(ProductBase):
    __slots__ = ('weight_kg',)
    def __init__(self, name, price, stock, weight_kg):
        super().__init__(name, price, stock)
        self.weight_kg = float(weight_kg)
    def shipping_cost(self): return round(self.weight_kg * 10.0, 2)

class Cart:
    def __init__(self): self.items = {}; self.discount = 0.0
    def add(self, product, qty=1): self.items[product] = self.items.get(product, 0) + qty
    def apply_discount(self, d): self.discount = d
    def total(self):
        items_cost = sum(p.price * q for p, q in self.items.items()) * (1 - self.discount)
        shipping = sum(p.shipping_cost() * q for p, q in self.items.items() if isinstance(p, PhysicalProduct))
        return round(items_cost + shipping, 2)

class Order:
    def __init__(self, cart):
        self.total = cart.total()
        self.status = "Placed"

if __name__ == '__main__':
    cart = Cart()
    cart.add(PhysicalProduct("Book", 20.0, 10, 0.5), 1)
    cart.apply_discount(0.1)
    order = Order(cart)
    print("Q138 Output:", {'total': cart.total(), 'order_status': order.status})
```

```python
# ==========================================
# 12. OBJECT-ORIENTED PROGRAMMING (Selected: Q139)
# ==========================================
# Question 139:
# Write a class DataClassValidator using __init_subclass__ so that any class inheriting from it and 
# defining a SCHEMA class variable (a dict of field->type) automatically validates that all required fields are 
# present and correctly typed on instantiation. Demonstrate with two model classes.
#
# Sample Input:  UserModel(user_id=1, username="alex")
# Sample Output: {'user_id': 1, 'username': 'alex'}

class DataClassValidator:
    SCHEMA = {}
    def __init_subclass__(cls, **kwargs):
        super().__init_subclass__(**kwargs)
        if not hasattr(cls, 'SCHEMA') or not isinstance(cls.SCHEMA, dict):
            raise AttributeError("Class must define dict 'SCHEMA'")

    def __init__(self, **kwargs):
        for field, typ in self.SCHEMA.items():
            if field not in kwargs or not isinstance(kwargs[field], typ):
                raise TypeError(f"Invalid or missing field: {field}")
            setattr(self, field, kwargs[field])

class UserModel(DataClassValidator):
    SCHEMA = {'user_id': int, 'username': str}

if __name__ == '__main__':
    user = UserModel(user_id=1, username="alex")
    print("Q139 Output:", user.__dict__)
```
