```python
# ==========================================
# 3. COMBINATORICS & PERMUTATIONS (Selected: Q104)
# ==========================================
# Question 104:
# Write a function that generates all permutations of a list using a loop-based approach (Heap's 
# algorithm). Do not use itertools.permutations. Verify the count against n!.
#
# Sample Input:  lst = [1, 2, 3]
# Sample Output: ([[1, 2, 3], [2, 1, 3], [3, 1, 2], [1, 3, 2], [2, 3, 1], [3, 2, 1]], True)
import math
def heap_permutations_iterative(lst):
    arr = lst.copy()
    result = [arr.copy()]
    n = len(lst)
    count = [0]*n
    i = 0
    while i < n:
        if count[i] < i:
            if i % 2 == 0:
                arr[0],arr[i] = arr[i],arr[0]
            else:
                arr[count[i]],arr[i] = arr[i],arr[count[i]]
            result.append(arr.copy())
            count[i] += 1
            i = 0
        else:
            count[i] = 0
            i += 1
    return result, len(result) == math.factorial(n)

if __name__ == '__main__':
    # Test Question 104
    print("Q104 Output:", heap_permutations_iterative([1, 2, 3]))
```

```python
# ==========================================
# 4. DISCRETE EVENT SIMULATION (Selected: Q105)
# ==========================================
# Question 105:
# Implement a simulation of a simple bank queue. Customers arrive at random intervals (1–5 minutes) 
# and each transaction takes a random time (2–8 minutes). Run the simulation for 60 minutes and report: 
# average wait time, maximum queue length, and total customers served.
#
# Sample Input:  simulation_time = 60, seed = 42
# Sample Output: {'avg_wait_time': 4.25, 'max_queue_length': 3, 'total_served': 12}

from collections import deque
import random
def simulate_bank_queue(simulation_time=60, seed=None):
    if seed is not None:
        random.seed(seed)
    queue = deque()
    curr_customer = None
    start_time= 0
    serviceend_time = 0

    total_wait = 0
    total_served = 0
    max_queue = 0

    next_arrival = random.randint(1,5)
    for time in range(simulation_time):

        if time == next_arrival:
            queue.append(time)
            next_arrival += random.randint(1,5)

        if curr_customer is not None and time >= serviceend_time:
            total_served += 1
            curr_customer = None

        if curr_customer is None and queue:
            arrival_time = queue.popleft()
            total_wait += time - arrival_time

            curr_customer = arrival_time
            serviceend_time = time + random.randint(2,8)
        max_queue = max(max_queue, len(queue))
    avg_wait = total_wait / total_served if total_served else 0

    return {
        "avg_wait_time": round(avg_wait, 2),
        "max_queue_length": max_queue,
        "total_served": total_served
    }
if __name__ == '__main__':
    # Test Question 105
    print("Q105 Output:", simulate_bank_queue(60, seed=42))
```

```python
# ==========================================
# 5. NESTED LOOPS & MATRIX FORMATTING (Selected: Q106)
# ==========================================
# Question 106:
# Write a program that prints a 10×10 multiplication table using nested loops, formatted so that all 
# columns are right-aligned to the width of the largest value. Then rotate the table 90 degrees clockwise and 
# print the rotated version.
#
# Sample Input:  size = 10
# Sample Output: {'original_formatted_str': '...', 'rotated_formatted_str': '...'}

import numpy as np

def formatted_multiplication_table(size=10):
    # Create multiplication table
    m = np.zeros((size, size), dtype=int)

    for i in range(size):
        for j in range(size):
            m[i][j] = (i + 1) * (j + 1)

    # Width of largest value
    width = len(str(size * size))

    # Format original table
    original = "\n".join(
        " ".join(f"{x:>{width}}" for x in row)
        for row in m
    )

    # Rotate 90° clockwise
    rotated_m = m.T
    rotated_m = rotated_m[:, ::-1]

    # Format rotated table
    rotated = "\n".join(
        " ".join(f"{x:>{width}}" for x in row)
        for row in rotated_m
    )

    return {
        "original_formatted_str": original,
        "rotated_formatted_str": rotated
    }


if __name__ == '__main__':
    output = formatted_multiplication_table(10)
    print("Q106 Output:")
    print(output["original_formatted_str"])
    print()
    print(output["rotated_formatted_str"])
```
