```python
# ==========================================
# 2. STACK-BASED PARSING (Selected: Q96)
# ==========================================
# Question 96:
# Write a function that takes a string containing nested parentheses, brackets, and braces and returns 
# True if they are all correctly balanced. For example, "{[()]}" is balanced but "{[(])}" is not. Use a stack (list). 
# Handle edge cases: empty string, single characters.
#
# Sample Input:  s = "{[()]}"
# Sample Output: True

def is_balanced_brackets(s):
    stack = []
    for i in s:
        if i in ["(","{","["]:
            stack.append(i)
        else:
            if not stack: return False
            top = stack.pop()
            if i == ")" and top != "(": return False
            elif i == "]" and top != "[": return False
            elif i == "}" and top != "{": return False
        # print(stack)
    return  stack == []

if __name__ == '__main__':
    # Test Question 96
    print("Q96 Output:", is_balanced_brackets("{[()]}"))
```

```python
# ==========================================
# 3. NUMBER CONVERSION (Selected: Q97)
# ==========================================
# Question 97:
# Write a function that takes a Roman numeral string (e.g., "MCMXCIV") and converts it to an integer. 
# Then write the inverse: integer to Roman numeral. Handle values from 1 to 3999.
#
# Sample Input:  roman_str = "MCMXCIV"
# Sample Output: (1994, 'MCMXCIV')

def roman_to_int(s):
    data = ["I","V","X","L","C","D","M"]
    value = [1,5,10,50,100,500,1000]
    val = 0
    for i in range(len(s)-1):
        if data.index(s[i]) < data.index(s[i+1]):
            val -= value[data.index(s[i])]
        else: val += value[data.index(s[i])]
    val += value[data.index(s[-1])]
    return val

def int_to_roman(num):
    roman_values = [
    (1000, "M"),
    (900, "CM"),
    (500, "D"),
    (400, "CD"),
    (100, "C"),
    (90, "XC"),
    (50, "L"),
    (40, "XL"),
    (10, "X"),
    (9, "IX"),
    (5, "V"),
    (4, "IV"),
    (1, "I")
    ]
    val = ""
    for value, symbol in roman_values:
        while num >= value:
            val += symbol
            num -= value

    return val

if __name__ == '__main__':
    # Test Question 97
    num_val = roman_to_int("MCMXCIV")
    roman_val = int_to_roman(num_val) if num_val is not None else None
    print("Q97 Output:", (num_val, roman_val))
```

```python
# ==========================================
# 4. EXPRESSION EVALUATION (Selected: Q98)
# ==========================================
# Question 98:
# Write a function that takes a simple arithmetic expression as a string (with +, -, *, / operators and 
# integer operands, no parentheses) and evaluates it correctly respecting operator precedence. Do not use 
# eval().
#
# Sample Input:  expr = "3 + 5 * 2 - 8 / 4"
# Sample Output: 11.0

def evaluate_expression(expr):
    stack = []
    op = '+'
    for i in expr.split():
        if i.isdigit():
            num = int(i)
            if op == "+": stack.append(num)
            elif op == "-": stack.append(-num)
            elif op == "*": stack.append(stack.pop()*num)
            elif op =="/": stack.append(stack.pop()/num)
        else: op = i
    return sum(stack)



if __name__ == '__main__':
    # Test Question 98
    print("Q98 Output:", evaluate_expression("3 + 5 * 2 - 8 / 4"))
```

```python
# ==========================================
# 5. DYNAMIC PROGRAMMING (Selected: Q99)
# ==========================================
# Question 99:
# Write a function longest_common_subsequence(s1, s2) that returns both the length and the actual 
# LCS string using dynamic programming. Verify with at least three test pairs.
#
# Sample Input:  s1 = "AGGTAB", s2 = "GXTXAYB"
# Sample Output: (4, 'GTAB')

def longest_common_subsequence(s1, s2):
    m = len(s1)
    n = len(s2)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(1, m + 1):
        for j in range(1, n + 1):

            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1

            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    # Build the actual LCS
    i = m
    j = n
    lcs = []

    while i > 0 and j > 0:

        if s1[i - 1] == s2[j - 1]:
            lcs.append(s1[i - 1])
            i -= 1
            j -= 1

        elif dp[i - 1][j] >= dp[i][j - 1]:
            i -= 1

        else:
            j -= 1

    lcs.reverse()

    return dp[m][n], ''.join(lcs)
if __name__ == '__main__':
    # Test Question 99
    print("Q99 Output:", longest_common_subsequence("AGGTAB", "GXTXAYB"))
```

```python
# ==========================================
# 6. TEXT PARSING (Selected: Q100)
# ==========================================
# Question 100:
# Write a function that takes a paragraph and returns all sentences that contain a given keyword 
# (case-insensitive), along with the sentence index (starting from 1). A sentence ends at ".","!", or "?". Do not use regex.
#
# Sample Input:  text = "Python is great. Learning it is fun! Are you ready?", keyword = "python"
# Sample Output: [(1, 'Python is great')]

def find_keyword_sentences(paragraph, keyword):
    data = []
    res = []
    s = ""
    for i in paragraph:
        if i not in ['.','?','!']:
            s += i
        else:
            data.append(s)
            s = ''
    for i in data:
        if keyword in i.strip().lower().split():
            res.append((data.index(i)+1,i))
    return res

if __name__ == '__main__':
    # Test Question 100
    sample_para = "Python is great. Learning it is fun! Are you ready?"
    print("Q100 Output:", find_keyword_sentences(sample_para, "python"))
```
