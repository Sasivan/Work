```python
# ==========================================
# 1. TAX CALCULATOR (Selected: Q107)
# ==========================================
# Question 107:
# Write a function calculate_income_tax(income, filing_status) that computes Indian income tax 
# under the new tax regime (FY 2024–25 slabs). Handle all slabs, the standard deduction, and surcharge for 
# incomes above ₹50L. Print a full tax breakdown table.
#
# Sample Input:  income = 1200000, filing_status = "individual"
# Sample Output: {'Taxable Income': 1125000.0, 'Slab Tax': 82500.0, 'Surcharge': 0.0, 'Cess (4%)': 3300.0, 'Total Tax': 85800.0}

def calculate_income_tax(income, filing_status="individual"):
    # Standard deduction
    standard_deduction = 50000
    taxable_income = max(0, income - standard_deduction)

    # FY 2024-25 new regime slabs
    if taxable_income <= 300000:
        slab_tax = 0

    elif taxable_income <= 600000:
        slab_tax = (taxable_income - 300000) * 0.05

    elif taxable_income <= 900000:
        slab_tax = 15000 + (taxable_income - 600000) * 0.10

    elif taxable_income <= 1200000:
        slab_tax = 45000 + (taxable_income - 900000) * 0.15

    elif taxable_income <= 1500000:
        slab_tax = 90000 + (taxable_income - 1200000) * 0.20

    else:
        slab_tax = 150000 + (taxable_income - 1500000) * 0.30

    # Surcharge
    if taxable_income > 5000000:
        surcharge = slab_tax * 0.10
    else:
        surcharge = 0

    # 4% Health & Education Cess
    cess = (slab_tax + surcharge) * 0.04

    total_tax = slab_tax + surcharge + cess

    return {
        "Taxable Income": taxable_income,
        "Slab Tax": slab_tax,
        "Surcharge": surcharge,
        "Cess (4%)": cess,
        "Total Tax": total_tax
    }


if __name__ == '__main__':
    print("Q107 Output:",
          calculate_income_tax(600000, "individual"))
```

```python
# ==========================================
# 2. RULE ENGINE (Selected: Q108)
# ==========================================
# Question 108:
# Write a rule-engine function evaluate_loan_eligibility(applicant: dict) -> (bool, list[str]) that checks: 
# age 21–65, CIBIL score >= 700, monthly income >= 3x EMI, existing EMIs <= 50% of income, and no defaults in 
# the last 2 years. Return True/False and a list of reason strings.
#
# Sample Input:  applicant = {'age': 30, 'cibil': 750, 'monthly_income': 100000, 'requested_emi': 20000, 'existing_emis': 20000, 'defaults_last_2yrs': 0}
# Sample Output: (True, ['All eligibility criteria met successfully.'])

def evaluate_loan_eligibility(applicant: dict):
    check = True
    if 21 > applicant["age"] > 65: check = False
    if applicant["cibil"] < 700: check = False
    if applicant["monthly_income"] < 3*applicant["requested_emi"]: check = False
    if applicant["existing_emis"] > 0.5*applicant["monthly_income"]: check = False
    if applicant["defaults_last_2yrs"] > 0: check = False
    return check

if __name__ == '__main__':
    # Test Question 108
    sample_applicant = {
        'age': 30,
        'cibil': 750,
        'monthly_income': 100000,
        'requested_emi': 20000,
        'existing_emis': 20000,
        'defaults_last_2yrs': 0
    }
    print("Q108 Output:", evaluate_loan_eligibility(sample_applicant))
```

```python
# ==========================================
# 3. DATE NORMALIZATION (Selected: Q109)
# ==========================================
# Question 109:
# Write a function that takes a date string in any of these formats: "DD/MM/YYYY", "MM-DD-YYYY", 
# "YYYY.MM.DD", "D Month YYYY" (e.g., "5 August 2026") and normalises it to "YYYY-MM-DD". Use only 
# string methods and conditional logic — no datetime module.
#
# Sample Input:  date_str = "5 August 2026"
# Sample Output: '2026-08-05'

def normalize_date_string(date_str):
    if '/' in date_str:
        d = date_str.split('/')
        return f"{d[-1]}-{d[1]}-{d[0]}"
    elif '-' in date_str:
        d = date_str.split('-')
        return f"{d[-1]}-{d[0]}-{d[1]}"
    elif '.' in date_str:
        d = date_str.split('.')
        return f"{d[0]}-{d[1]}-{d[2]}"
    else:
        d = date_str.split(' ')
        m = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]
        return f"{d[-1]}-{m.index(d[1])+1:02}-{int(d[0]):02}"
if __name__ == '__main__':
    # Test Question 109
    print("Q109 Output:", normalize_date_string("5 August 2026"))
```
