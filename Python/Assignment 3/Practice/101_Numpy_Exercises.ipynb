**49. Compute the Row-Wise Counts of All Possible Values in an Array**

**Task:** Compute the counts of unique values row-wise for the array `arr`.`arr = np.random.randint(1, 11, size=(6, 10))`

**Your Solution:**

````python
# Write your solution here
````

**Solution:**

````python
import numpy as np
np.random.seed(100)
Z = np.random.randint(10, size=10)
print('array:', Z)

def moving_average(a, n=3):
    ret = np.cumsum(a, dtype=float)
    ret[n:] = ret[n:] - ret[:-n]
    return ret[n - 1:] / n

print(moving_average(Z, n=3).round(2))
````

**Solution:**

````python
import numpy as np
arr = np.arange(15)

def gen_strides(a, stride_len=5, window_len=5):
    n_strides = ((a.size - window_len) // stride_len) + 1
    return np.array([a[s:(s + window_len)] for s in np.arange(0, n_strides * stride_len, stride_len)])

print(gen_strides(arr, stride_len=2, window_len=4))
````

**Task:** Perform PCA on the dataset `X` (rows = observations, columns = features) by computing the covariance matrix, then its eigenvalues/eigenvectors, and project `X` onto the top principal component.`X = np.array([[2.5, 2.4], [0.5, 0.7], [2.2, 2.9], [1.9, 2.2], [3.1, 3.0]])`

**Your Solution:**

````python
# Write your solution here
````

**Solution:**

````python
import numpy as np
X = np.array([[2.5, 2.4], [0.5, 0.7], [2.2, 2.9], [1.9, 2.2], [3.1, 3.0]])
X_centered = X - X.mean(axis=0)
cov_mat = np.cov(X_centered, rowvar=False)
eig_vals, eig_vecs = np.linalg.eig(cov_mat)
order = np.argsort(eig_vals)[::-1]
top_vec = eig_vecs[:, order[0]]
projected = X_centered @ top_vec
print('Eigenvalues:', eig_vals)
print('Projected onto PC1:', projected)
````

**99. Compute 2D Sliding Windows Over a Matrix**

**Task:** Extract all contiguous 2x2 sliding windows (blocks) from the 2D array `a`.`a = np.arange(16).reshape(4, 4)`

**Your Solution:**

````python
# Write your solution here
````

**Solution:**

````python
import numpy as np
a = np.arange(16).reshape(4, 4)
window = 2
windows = []
for i in range(a.shape[0] - window + 1):
    for j in range(a.shape[1] - window + 1):
        windows.append(a[i:i + window, j:j + window])
result = np.array(windows)
print(result)
````

**101. Compute a Pairwise Euclidean Distance Matrix Without Loops**

**Task:** Compute the pairwise euclidean distance matrix between all rows of `X` without using explicit Python loops.`X = np.array([[0, 0], [3, 4], [6, 8]])`

**Your Solution:**

````python
# Write your solution here
````

**Solution:**

````python
import numpy as np
X = np.array([[0, 0], [3, 4], [6, 8]])
sq_norms = np.sum(X**2, axis=1)
dist_sq = sq_norms[:, None] + sq_norms[None, :] - 2 * X @ X.T
dist_sq = np.maximum(dist_sq, 0)
dist = np.sqrt(dist_sq)
print(dist)
````
