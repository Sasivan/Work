```python
# Q15: Vectorize a Scalar Function to Work on Arrays
# Difficulty: L3

import numpy as np

def maxx(x, y):
    """Get the maximum of two items"""
    if x >= y:
        return x
    else:
        return y

pair = np.vectorize(maxx)

a = np.array([5, 7, 9, 8, 6, 4, 5])
b = np.array([6, 3, 4, 8, 9, 7, 1])

result = pair(a,b)

print("Q15 Result:", result)
```
