<h1>Question 1</h1>

```python
import mysql.connector

def fetch(cursor,chunk_size = 2000):
    while True:
        results = cursor.fetchmany(chunk_size)
        if not results:
            break
        for row in results:
            yield row

def main():
    db_config = {
        "host":"localhost",
        "user":"root",
        "password":"sasivan",
        "database":"employees"
    }
    try:
        conn = mysql.connector.connect(**db_config)
        cursor = conn.cursor(dictionary = True)
        print("Executing -")
        cursor.execute("select emp_no,salary,from_date from salaries;")

        salary_stream = fetch(cursor,chunk_size = 2000)

        count = 0
        total_salary = 0
        print("Processing records -")
        for record in salary_stream:
            count += 1
            total_salary += record["salary"]
        print(f"Record - {count} \nAvg salary - ${total_salary/count:.2f}")
    except mysql.Connector.Error as err:
        print(f"Error - {err}")
    finally:
        if 'conn' in locals() and conn.is_connected():
            cursor.close()
            conn.close()
            print("Exit")
if __name__ == "__main__":
    main()
```

<h1>Question 3</h1>

```python
def remove_duplicate(seq):
    seen = set()
    result = []
    for i in seq:
        if i not in seen:
            seen.add(i)
            result.append(i)
    return result

original_list = [12, 24, 35, 24, 88, 120, 155, 88, 120, 155]

res = remove_duplicate(original_list)
print(f"Original - {original_list}")
print(f"Unique - {res}")
```

<h1>Question 5</h1>

```python
def binary_search(arr,t):
    l = 0
    r = len(arr) - 1
    while l <= r:
        m = (r+l)//2
        if arr[m] > t:  r = m - 1
        elif arr[m] < t:    l = m + 1
        else: return m
    return -1
if __name__ == "__main__":
    sorted_list = [12, 24, 35, 88, 120, 155]
    search_item = 88

    result = binary_search(sorted_list, search_item)
    print(result)
```

```python
import os

def tree(path, space=""):
    files = os.listdir(path)

    for file in files:
        print(space + "|--", file)

        full_path = os.path.join(path, file)

        if os.path.isdir(full_path):
            tree(full_path, space + "    ")

folder = "C:\Work\Python\Assignments"
tree(folder)
```

<h1>Question 10</h1>

```python
class ReverseIter:
    def __init__(self,data):
        self.data = data
        self.ind = len(data) - 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.ind >= 0:
            item = self.data[self.ind]
            self.ind -= 1
            return item
        else:
            raise StopIteration
if __name__ == "__main__":
    res = ReverseIter([10, 20, 30, 40, 50])
    print(*res)
```
