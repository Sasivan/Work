```python
import mysql.connector

def fetch_records_in_chunks(query, batch_size=5000):
    conn = mysql.connector.connect(host='localhost', user='root', password='sasivan', database='test_db')
    cursor = conn.cursor(dictionary=True)
    cursor.execute(query)

    while True:
        records = cursor.fetchmany(batch_size)
        if not records:
            break
        for record in records:
            yield record  
        
    cursor.close()
    conn.close()
```

```python
import sys
import urllib.request
import re

def strip_html_tags(url):
    try:
        # Download HTML
        response = urllib.request.urlopen(url)
        html = response.read().decode('utf-8')
        # Regex to strip tags
        clean_text = re.sub(r'<[^>]+>', '', html)
        print(clean_text)
    except Exception as e:
        print(f"Error fetching URL: {e}")

if __name__ == "__main__":
    if len(sys.argv) > 1:
        strip_html_tags(sys.argv[1])
```

```python
from flask import Flask, request, jsonify

app = Flask(__name__)

customers = []
products = []
orders = []

@app.route('/customers', methods=['GET', 'POST', 'PUT'])
def customer():
    if request.method == 'POST':
        customers.append(request.json)
        return jsonify({"message": "Customer added"})
    if request.method == 'PUT':
        customers[0] = request.json
        return jsonify({"message": "Customer updated"})
    return jsonify(customers)

@app.route('/products', methods=['GET', 'POST', 'PUT'])
def product():
    if request.method == 'POST':
        products.append(request.json)
        return jsonify({"message": "Product added"})
    if request.method == 'PUT':
        products[0] = request.json
        return jsonify({"message": "Product updated"})
    return jsonify(products)

@app.route('/orders', methods=['GET', 'POST', 'PUT'])
def order():
    if request.method == 'POST':
        orders.append(request.json)
        return jsonify({"message": "Order added"})
    if request.method == 'PUT':
        orders[0] = request.json
        return jsonify({"message": "Order updated"})
    return jsonify(orders)

app.run(debug=True)
```

```python
import pandas as pd

df = pd.read_csv("survey_results.csv")

# 1. Average age
print(df["Age1stCode"].astype(float).mean())

# 2. Python percentage by country
df["Python"] = df["LanguageWorkedWith"].fillna("").str.contains("Python")

result = df.groupby("Country")["Python"].mean() * 100
print(result)

# 3. Salary by continent
country_continent = {
    "India": "Asia",
    "United States": "North America",
    "United Kingdom": "Europe",
    "Germany": "Europe",
    "Australia": "Oceania"
}

df["Continent"] = df["Country"].map(country_continent)

print(df.groupby("Continent")["ConvertedComp"].mean())

# 4. Gender cleaning
df["Gender"] = df["Gender"].apply(
    lambda x: "MAN" if x == "Man"
    else ("WOMAN" if x == "Woman" else "OTHERS")
)

print(df["Gender"].value_counts())
```

```python
from flask import Flask, request
import subprocess

app = Flask(__name__)

@app.route("/")
def home():
    return """
    <textarea id="note"></textarea>

    <script>
    let timer;

    document.getElementById("note").addEventListener("keyup", function() {
        clearTimeout(timer);

        timer = setTimeout(() => {
            fetch("/save", {
                method: "POST",
                body: this.value
            });
        }, 1000);
    });
    </script>
    """

@app.route("/save", methods=["POST"])
def save():
    text = request.data.decode()

    with open("note.txt", "w") as f:
        f.write(text)

    subprocess.run(["git", "add", "."])
    subprocess.run(["git", "commit", "-m", "Auto-save"])

    return "Saved"

app.run(debug=True)
```

```python
import re
import json

with open("sample.sql", "r") as f:
    sql = f.read()

procedures = re.findall(
    r"CREATE PROCEDURE\s+([A-Za-z0-9_]+)",
    sql,
    re.I
)

tables = re.findall(
    r"(?:FROM|JOIN|INTO|UPDATE)\s+([A-Za-z0-9_]+)",
    sql,
    re.I
)

result = {
    "procedures": procedures,
    "tables": list(set(tables))
}

with open("sample_stored_procedure.json", "w") as f:
    json.dump(result, f, indent=4)

print(result)
```

```python
import pandas as pd
from sqlalchemy import create_engine
import re

# Read JSON
df = pd.read_json("sample_data_for_assignment.json")

# MySQL connection
engine = create_engine(
    "mysql+pymysql://root:password@localhost/test"
)

# Insert into MySQL
df.to_sql("json_to_sql_table", engine, if_exists="replace", index=False)

# Read back
df = pd.read_sql("SELECT * FROM json_to_sql_table", engine)

# Change email domain
df["email"] = df["email"].apply(
    lambda x: x.split("@")[0] + "@gmail.com"
)

# Clean ZIP
df["postalZip"] = (
    df["postalZip"]
    .astype(str)
    .str.replace(r"\D", "", regex=True)
    .astype(int)
)

# Phone encoding
def encode_phone(phone):
    digits = re.sub(r"\D", "", str(phone))
    result = ""

    for i in range(0, len(digits) - 1, 2):
        value = int(digits[i:i+2])

        if value < 65:
            result += "O"
        else:
            result += chr(value)

    return result

df["coded_phone_number"] = df["phone"].apply(encode_phone)

print(df)
```
